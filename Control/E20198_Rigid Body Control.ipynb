**Task 1: Derivation of the Mathematical Model**

Let $\mathbf{b}$ be an orthonormal frame with centre $O$ fixed to the body of the twin rotors including the beam connecting them and $\mathbf{c}$ be another orthonormal frame with centre $O$ fixed to the upright support which the twin rotors assembly is pivoted to as shown in Figure 1a.

The twin rotors can move about its pivot in such a way that $\mathbf{b}_1 \equiv \mathbf{c}_1$. The frame $\text{c}$ or the upright support rotates about the axis $\mathbf{c}_3 \equiv \mathbf{e}_3$, where $\mathbf{e}$ is the Earth fixed inertial frame shown in Figure 1b below.

<center>
<div style="text-align: center;">

<img src="https://github.com/chamudithamk/ME422-B2-Lab-Group/blob/main/Control/Images/RotorTop.png?raw=1" width="40%" style="display: block; margin: 0 auto;">

The frame $\mathbf{c}$ is related to frame $\mathbf{e}$ in such a way that:

$$
\mathbf{c}=\mathbf{e}\,R_3(\theta)
$$

$$
R_3(\theta)=\begin{bmatrix}
\cos{\theta}&-\sin{\theta}&0\\
\sin{\theta}&\cos{\theta}&0\\
0&0&1
\end{bmatrix}$$

The frame $\mathbf{b}$ is related to frame $\mathbf{c}$ in such a way that:

$$
\mathbf{b}=\mathbf{c}\,R_1(\phi)
$$

$$
R_1(\phi)=\begin{bmatrix}
1&0&0\\
0&\cos{\phi}&-\sin{\phi}\\
0&\sin{\phi}&\cos{\phi}\end{bmatrix}
$$

By combining above results, the kinematics of the purely rotating body can be parameterised to an observer in the Earth frame using Euler angle representation.

$$
\mathbf{b}=\mathbf{c}R_1(\phi)=\mathbf{e}\underbrace{R_3(\theta)R_1(\phi)}_{R} =\mathbf{e}R
$$

where

$$R = R_3 R_1 = \begin{bmatrix}
\cos{\theta}&-\cos{\phi}\sin{\theta}&\sin{\phi}\sin{\theta}\\
\sin{\theta}&\cos{\phi}\cos{\theta}&-\sin{\phi}\cos{\theta}\\
0&\sin{\phi}&\cos{\phi}\end{bmatrix}$$

**Part I: Derivation of Angular Velocity**

By definition, it can be said that skew-symmetric matrix terms $\hat{\omega}$ and $\hat{\Omega}$ of twin rotor $\mathbf{b}$ w.r.t to $\mathbf{c}$ and $\mathbf{c}$ w.r.t. the Earth fixed frame $\mathbf{e}$ in domain $\mathfrak{so}(3)$ is given by the relationship:

$$\dot{R_1} = \hat{\omega}_1 R_1 = R_1 \hat{\Omega}_1$$
$$\dot{R_3} = \hat{\omega}_3 R_3 = R_3 \hat{\Omega}_3$$

$$
\hat{\omega}_1 = \dot{R}_1 R_1^T = \begin{bmatrix}
0&0&0\\
0&0&-\dot{\phi}\\
0&\dot{\phi}&0\end{bmatrix}
$$

$$
\hat{\omega}_3 = \dot{R}_3 R_3^T =\begin{bmatrix}
0&-\dot{\theta}&0\\
\dot{\theta}&0&0\\
0&0&0
\end{bmatrix}
$$

Then, to derive a relationship for skew-symmetric matrix $\hat{\omega}$ of twin rotor, $\mathbf{b}$ w.r.t. Earth frame $\mathbf{e}$,

$$
R = R_3(\theta)R_1(\phi)
$$

Differentiating w.r.t time:

$$
\dot{R} = \dot{R}_3R_1+R_3\dot{R}_1 $$
$$
\dot{R} = (\hat{\omega}_3 R_3) R_1+R_3(\hat{\omega}_1 R_1) $$

Since $R_3^T R_3 = I$,

$$\dot{R} = \hat{\omega}_3 (R_3 R_1) + R_3 \hat{\omega}_1(R_3^T R_3)R_1 $$
$$\dot{R} = \hat{\omega}_3 R + (R_3 \hat{\omega}_1 R_3^T)R$$
$$\dot{R} = \underbrace{(\hat{\omega}_3 + R_3 \hat{\omega}_1 R_3^T)}_{\hat{\omega}}R$$

It can be shown that:

$$\dot{R} = \hat{\omega}R$$

where

$$ \hat{\omega} = \hat{\omega}_3 + R_3 \hat{\omega}_1 R_3^T = \begin{bmatrix}
0&-\dot{\theta}&\dot{\phi}\sin{\theta}\\
\dot{\theta}&0&-\dot{\phi}\cos{\theta}\\
-\dot{\phi}\sin{\theta}&\dot{\phi}\cos{\theta}&0
\end{bmatrix}$$

Thus, there exists a skew-symmetric matrix $\hat{\omega} \in \mathfrak{so}(3)$ for spatial angular velocity of twin rotor $\mathbf{b}$ with respect to Earth frame $\mathbf{e}$ corresponding to rototion matrix $R \in SO(3)$ defined above.

The spatial angular velocity $\omega \in \mathbb{R}^3$ for the twin rotor is then given by,

$$\omega = \begin{bmatrix}
\dot{\phi} \cos{\theta} \\
\dot{\phi} \sin{\theta} \\
\dot{\theta}
\end{bmatrix}$$

The body angular velocity $\Omega \in \mathbb{R}^3$ for the twin rotor is given by,

$$
\Omega = R^T \omega = \begin{bmatrix}
\dot{\phi} \\
\dot{\theta} \sin{\phi} \\
\dot{\theta} \cos{\phi}
\end{bmatrix}
$$

Here, $\Omega$ is the angular velocity we measure through the sensors whereas $\omega$ is the angular velocity we observe.

**Part II: Derivation of Torque Equation**

Considering $O$ to be the centre of mass of the twin rotor system, let $\mathbb{I}$ be the body inertia tensor and $\mathbb{I}^R = R \mathbb{I} R^T$ be the locked inertia tensor in $\mathbf{e}$ frame. Similarly, let $\pi$ be the spatial angular momentum and $\Pi = R^T \pi$ be the body angular momentum.

Given that,

$$
\omega =(\mathbb{I}^R)^{-1}\pi
$$

By Newton's 2nd law, the rate of change of angular momentum:

$$
\dot{\pi} = \tau_e + \tau_u
$$

where $\tau_e$ is the reaction torque (unmanipulatable) due to holonomic constraints and $\tau_u$ is the actuator torques (manipulatable) by propellers.

Let $\alpha$ and $\beta$ be the offset angles of propellers fans and $u_1$ and $u_2$ be the effective input torques by each propeller. Then, the actuator torque is given by,

$$
T_u = \begin{bmatrix}
\cos{\alpha} & -\cos{\beta}\\
\sin{\alpha} & -\sin{\beta}
\end{bmatrix}
\begin{bmatrix}
u_1\\ u_2
\end{bmatrix}
$$

Note that from Figure 1a, one fan controls the rotation about $\mathbf{b_1}$ and the other fan controls the rotation about $\mathbf{b_3}$. Then, the torque generated in the body frame can written as,

$$
T_u = \begin{bmatrix}
1 & 0\\
0 & 0\\
0 & 1
\end{bmatrix}
\begin{bmatrix}
\cos{\alpha} & -\cos{\beta}\\
\sin{\alpha} & -\sin{\beta}
\end{bmatrix}
\begin{bmatrix}
u_1\\ u_2
\end{bmatrix}
$$

For the actuator torque is spatial frame $\mathbf{e}$,

$$
\tau_u =R\begin{bmatrix}
1 & 0\\
0 & 0\\
0 & 1
\end{bmatrix}
\begin{bmatrix}
\cos{\alpha} & -\cos{\beta}\\
\sin{\alpha} & -\sin{\beta}
\end{bmatrix}
\begin{bmatrix}
u_1\\ u_2
\end{bmatrix}$$

Since the two fans are $90^o$ apart, take $\alpha = 0$ and $\beta = \frac{\pi}{2}$.

$$
\tau_u =R\begin{bmatrix}
1 & 0\\
0 & 0\\
0 & 1
\end{bmatrix}
\begin{bmatrix}
1 & 0\\
0 & -1
\end{bmatrix}
\begin{bmatrix}
u_1\\ u_2
\end{bmatrix} = R\begin{bmatrix}
u_1\\
0\\
-u_2
\end{bmatrix}$$

$$
\tau_u = RT_u
$$


From Euler's Rigid Body equations, it can be written that the resultant torque acting on the body due to rotation is given by,

$$
T_e = \Omega \times \mathbb{I}\Omega+\mathbb{I}\dot{\Omega}
$$

Note that from the body's perspective, twin rotor system is allowed to rotate about $\mathbf{b_1}$ and $\mathbf{b_3}$ and is restricted to rotate about $\mathbf{b_2}$, or in other words about the rod that connects two propellers.

$$
T_e = \begin{bmatrix}0\\1\\0
\end{bmatrix} (\Omega \times \mathbb{I}\Omega+\mathbb{I}\dot{\Omega})
$$

For the holonomic constriant torque in spatial frame $\mathbf{e}$,

$$
\tau_e = R T_e
$$

Note that eventhough no torque is actuated about $\mathbf{b_2}$ in $\tau_u$, gyroscopic effects can induce unbalanced torques about that axis which needs to be compensated through a reaction $\tau_e$.

**Task 3: Derivation and Simulation of a PID Controller for the Tracking Problem using Error Dynamics**

Let $(o_r,R_r) \in SE(3)$ be the reference configuration of the twin rotor system. But since the centre of mass of the system is fixed at the pivot, $o_r \equiv o$  and thus, we can only focus on the dynamics of rotation $R$.

To ensure asymptotic tracking of the sufficiently smooth desired trajectory $R_r$, let us define the configuration error as $R_e \in SO(3)$ such that,

The left invariant error or the error related to body frame:

$$
R R_e = R_r \\
R_e = R^T R_r
$$

The right invariant error or the error related to Earth frame:

$$
R_e R = R_r \\
R_e = R_r R^T
$$

Since we defined our equation of motion in $\mathbf{e}$ frame, we'll use the latter in our calculations.



**Part I: Derivation of Tracking Error Dynamics**

Let $\omega_r$ be the spatial angular velocity of the desired trajectory such that $\widehat{\omega}_r = \dot{R}_rR_r^T$.

For the **error in spatial angular velocity**,

$$
\widehat{\omega}_e \triangleq \dot{R}_eR_e^T $$

$$
\widehat{\omega}_e = (\dot{R_r}R^T + R_r \dot{R}^T)R_e^T $$
$$
\widehat{\omega}_e = \dot{R_r}(R^T R_e^T) + (R_e R)  \dot{R}^T R_e^T $$
$$
\widehat{\omega}_e = \dot{R_r}R_r^T + R_e (R \dot{R}^T) R_e^T $$

From $RR^T = I \Rightarrow R\dot{R}^T + \dot{R}R^T = 0$

$$
\widehat{\omega}_e = \dot{R_r}R_r^T + R_e (-\dot{R} R^T) R_e^T $$

$$
\widehat{\omega}_e = \widehat{\omega}_r  - R_e \widehat{\omega} R_e^T $$

$$
\widehat{\omega}_e = \widehat{\omega}_r  - \widehat{R_e \omega} $$

Hence,

$$ \omega_e = \omega_r - R_e \omega $$


For the **error in spatial angular momentum**,

In line with the system define the reference spatial angular momentum $\pi_r\triangleq R_r\mathbb{I}R_r^T\omega_r$. Then, by defining the error in spatial angular momentum to be:

$$
\pi_e\triangleq R\mathbb{I}R_r^T\omega_e$$

$$
\pi_e = R\mathbb{I}R_r^T(\omega_r-R_e\omega) $$

$$
\pi_e = (R_e^T R_r)\mathbb{I}R_r^T\omega_r-R\mathbb{I}R^T\omega $$

$$
\pi_e =R_e^T\pi_r-\pi $$

Differentiating w.r.t time:

$$
\dot{\pi}_e = \dot{R_e^T}\pi_r + R_e^T\dot{\pi}_r - \dot{\pi} $$

$$
\dot{\pi}_e = ({R}_e^T \widehat{\omega}_e^T)\pi_r + R_e^T\dot{\pi}_r - \dot{\pi} $$

$$
\dot{\pi}_e = {R}_e^T (-\widehat{\omega}_e)\pi_r + R_e^T\dot{\pi}_r - \dot{\pi} $$

$$
\dot{\pi}_e = R_e^T(\dot{\pi}_r -\widehat{\omega}_e\pi_r)-\dot{\pi} $$

$$
\dot{\pi}_e = R_e^T(\dot{\pi}_r-\omega_e\times \pi_r)-\dot{\pi} $$

From $\pi_r = R_r \Pi_r$,

$$
\dot{\pi}_e = R_e^T(\dot{R}_r \Pi_r + R_r \dot{\Pi}_r-\omega_e\times \pi_r)-\dot{\pi} $$

$$
\dot{\pi}_e = R_e^T(\widehat{\omega}_r{R}_r \Pi_r + R_r \dot{\Pi}_r-\omega_e\times \pi_r)-\dot{\pi} $$

$$
\dot{\pi}_e = R_e^T(R_r \dot{\Pi}_r+ \widehat{\omega}_r \pi_r-\omega_e\times \pi_r)-\dot{\pi} $$

$$
\dot{\pi}_e = R_e^T(R_r\dot{\Pi}_r+(\omega_r-\omega_e)\times \pi_r)-\dot{\pi}$$

$$
\dot{\pi}_e = (R\dot{\Pi}_r+\omega\times \pi_r)-\dot{\pi}
$$

Thus, we have the error dynamics:
$$
\dot{R}_e=\widehat{\omega}_eR_e
$$
$$
\dot{\pi}_e=(R\dot{\Pi}_r+\omega\times \pi_r)-\tau_u-\tau_e
$$
