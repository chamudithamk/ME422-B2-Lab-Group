The reference frame $\mathbf{c}$ is related to the Earth–fixed inertial frame $\mathbf{e}$ through a rotation about the third axis by an angle $\theta$:

$$
\mathbf{c} = \mathbf{e} R_3(\theta)
$$

where the rotation matrix about the third axis is

$$
R_3(\theta)=
\begin{bmatrix}
\cos{\theta} & -\sin{\theta} & 0 \\
\sin{\theta} & \cos{\theta} & 0 \\
0 & 0 & 1
\end{bmatrix}
$$

Similarly, the frame $\mathbf{b}$ is obtained from frame $\mathbf{c}$ by a rotation about the first axis by an angle $\phi$:

$$
\mathbf{b} = \mathbf{c} R_1(\phi)
$$

where the rotation matrix about the first axis is

$$
R_1(\phi)=
\begin{bmatrix}
1 & 0 & 0 \\
0 & \cos{\phi} & -\sin{\phi} \\
0 & \sin{\phi} & \cos{\phi}
\end{bmatrix}
$$

Using the above relationships, the kinematics of the rigid body undergoing pure rotation can be described with respect to an observer in the Earth frame using the Euler angle representation.

$$
\mathbf{b} = \mathbf{c}R_1(\phi) = \mathbf{e}\underbrace{R_3(\theta)R_1(\phi)}_{R} = \mathbf{e}R
$$

where the combined rotation matrix is

$$
R = R_3 R_1 =
\begin{bmatrix}
\cos{\theta} & -\cos{\phi}\sin{\theta} & \sin{\phi}\sin{\theta} \\
\sin{\theta} & \cos{\phi}\cos{\theta} & -\sin{\phi}\cos{\theta} \\
0 & \sin{\phi} & \cos{\phi}
\end{bmatrix}
$$

**Derivation of Angular Velocity**

By definition, the skew–symmetric matrices $\hat{\omega}$ and $\hat{\Omega}$ associated with the twin rotor frame $\mathbf{b}$ relative to frame $\mathbf{c}$, and frame $\mathbf{c}$ relative to the Earth–fixed frame $\mathbf{e}$, belong to the Lie algebra $\mathfrak{so}(3)$. These matrices satisfy the following relationships:

$$
\dot{R}_1 = \hat{\omega}_1 R_1 = R_1 \hat{\Omega}_1
$$

$$
\dot{R}_3 = \hat{\omega}_3 R_3 = R_3 \hat{\Omega}_3
$$

The corresponding skew–symmetric matrices are given by

$$
\hat{\omega}_1 = \dot{R}_1 R_1^{T} =
\begin{bmatrix}
0 & 0 & 0 \\
0 & 0 & -\dot{\phi} \\
0 & \dot{\phi} & 0
\end{bmatrix}
$$

$$
\hat{\omega}_3 = \dot{R}_3 R_3^{T} =
\begin{bmatrix}
0 & -\dot{\theta} & 0 \\
\dot{\theta} & 0 & 0 \\
0 & 0 & 0
\end{bmatrix}
$$

To obtain the skew–symmetric matrix $\hat{\omega}$ of the twin rotor frame $\mathbf{b}$ with respect to the Earth frame $\mathbf{e}$, consider the rotation matrix

$$
R = R_3(\theta)R_1(\phi)
$$

Taking the time derivative,

$$
\dot{R} = \dot{R}_3R_1 + R_3\dot{R}_1
$$

Substituting the skew–symmetric representations of $\dot{R}_3$ and $\dot{R}_1$,

$$
\dot{R} = (\hat{\omega}_3 R_3)R_1 + R_3(\hat{\omega}_1 R_1)
$$

Since $R_3^{T}R_3 = I$,

$$
\dot{R} = \hat{\omega}_3 (R_3R_1) + R_3 \hat{\omega}_1 (R_3^{T}R_3) R_1
$$

$$
\dot{R} = \hat{\omega}_3 R + (R_3 \hat{\omega}_1 R_3^{T})R
$$

Thus,

$$
\dot{R} =
\underbrace{(\hat{\omega}_3 + R_3 \hat{\omega}_1 R_3^{T})}_{\hat{\omega}} R
$$

Therefore,

$$
\dot{R} = \hat{\omega}R
$$

where the resulting skew–symmetric matrix is

$$
\hat{\omega} = \hat{\omega}_3 + R_3 \hat{\omega}_1 R_3^{T} =
\begin{bmatrix}
0 & -\dot{\theta} & \dot{\phi}\sin{\theta} \\
\dot{\theta} & 0 & -\dot{\phi}\cos{\theta} \\
-\dot{\phi}\sin{\theta} & \dot{\phi}\cos{\theta} & 0
\end{bmatrix}
$$

Therefore, there exists a skew–symmetric matrix $\hat{\omega} \in \mathfrak{so}(3)$ representing the spatial angular velocity of the twin rotor frame $\mathbf{b}$ relative to the Earth frame $\mathbf{e}$, corresponding to the rotation matrix $R \in SO(3)$ defined previously.

The spatial angular velocity $\omega \in \mathbb{R}^3$ of the twin rotor system is expressed as

$$
\omega =
\begin{bmatrix}
\dot{\phi} \cos{\theta} \\
\dot{\phi} \sin{\theta} \\
\dot{\theta}
\end{bmatrix}
$$

The body angular velocity $\Omega \in \mathbb{R}^3$ of the twin rotor is obtained as

$$
\Omega = R^{T}\omega =
\begin{bmatrix}
\dot{\phi} \\
\dot{\theta}\sin{\phi} \\
\dot{\theta}\cos{\phi}
\end{bmatrix}
$$

Here, $\Omega$ represents the angular velocity measured by the onboard sensors, whereas $\omega$ corresponds to the angular velocity observed in the Earth frame.

**Derivation of Torque Equation**

Assume that the point $O$ represents the centre of mass of the twin rotor system. Let $\mathbb{I}$ denote the inertia tensor expressed in the body frame, and let $\mathbb{I}^R = R \mathbb{I} R^T$ represent the corresponding locked inertia tensor expressed in the Earth frame $\mathbf{e}$.

Similarly, let $\pi$ denote the spatial angular momentum and $\Pi = R^T \pi$ denote the angular momentum expressed in the body frame.

Given that

$$
\omega = (\mathbb{I}^R)^{-1}\pi
$$

According to Newton’s second law, the time derivative of angular momentum is

$$
\dot{\pi} = \tau_e + \tau_u
$$

where $\tau_e$ represents the reaction torque arising from holonomic constraints (which cannot be controlled), and $\tau_u$ denotes the actuator torques generated by the propellers (which are controllable).

Let $\alpha$ and $\beta$ be the offset angles of the two propeller fans, and let $u_1$ and $u_2$ represent the effective input torques produced by each propeller. The actuator torque can then be written as

$$
T_u =
\begin{bmatrix}
\cos{\alpha} & -\cos{\beta} \\
\sin{\alpha} & -\sin{\beta}
\end{bmatrix}
\begin{bmatrix}
u_1 \\
u_2
\end{bmatrix}
$$

From Figure 1a, one propeller controls rotation about $\mathbf{b}_1$ while the other controls rotation about $\mathbf{b}_3$. Therefore, the torque generated in the body frame can be expressed as

$$
T_u =
\begin{bmatrix}
1 & 0 \\
0 & 0 \\
0 & 1
\end{bmatrix}
\begin{bmatrix}
\cos{\alpha} & -\cos{\beta} \\
\sin{\alpha} & -\sin{\beta}
\end{bmatrix}
\begin{bmatrix}
u_1 \\
u_2
\end{bmatrix}
$$

The actuator torque expressed in the spatial frame $\mathbf{e}$ becomes

$$
\tau_u =
R
\begin{bmatrix}
1 & 0 \\
0 & 0 \\
0 & 1
\end{bmatrix}
\begin{bmatrix}
\cos{\alpha} & -\cos{\beta} \\
\sin{\alpha} & -\sin{\beta}
\end{bmatrix}
\begin{bmatrix}
u_1 \\
u_2
\end{bmatrix}
$$

Since the two fans are positioned $90^\circ$ apart, we take $\alpha = 0$ and $\beta = \frac{\pi}{2}$. Hence,

$$
\tau_u =
R
\begin{bmatrix}
1 & 0 \\
0 & 0 \\
0 & 1
\end{bmatrix}
\begin{bmatrix}
1 & 0 \\
0 & -1
\end{bmatrix}
\begin{bmatrix}
u_1 \\
u_2
\end{bmatrix}
=
R
\begin{bmatrix}
u_1 \\
0 \\
- u_2
\end{bmatrix}
$$

Therefore,

$$
\tau_u = R T_u
$$

**Task 2: Simulation of the Motion of the System for Different Inputs**

From the body frame $\mathbf{b}$, the quantities $\mathbb{I}_1$, $\mathbb{I}_2$, and $\mathbb{I}_3$ represent the moments of inertia associated with pitch, roll, and yaw motions, respectively.

Since the rolling motion is constrained in the experimental setup, the inertia term $\mathbb{I}_2$ does not play a role in the dynamics and can be neglected. Furthermore, it is observed that $\mathbb{I}_3 > \mathbb{I}_1$, because the yaw motion involves additional components (such as the upright support), whereas the pitch motion primarily involves only the beam connecting the two propellers.

```python
import numpy as np
import plotly.graph_objects as go
from scipy.integrate import solve_ivp

# ====================== System & Physical Parameters =========================
# Principal moments of inertia (assumed values)
I_x, I_y, I_z = 1.0, 2.0, 1.5

# ====================== Input Signal Definition =============================
def fetch_input_signals(time):
    # --- Enable ONLY ONE scenario at a time ---

    # signals = [0.5 * np.sin(2 * time), -0.2 * np.cos(time)]  # Frequency response
    # signals = [1.0, 1.0]                                    # Step response
    # signals = [1.0, 0.0]                                    # Pitch test
    signals = [0.0, 1.0]                                      # Yaw test

    return signals

# ====================== Actuator Torque Functions ===========================
# These represent the control torques to be determined in the experiment
def torque_u1(time):
    return fetch_input_signals(time)[0]

def torque_u2(time):
    return fetch_input_signals(time)[1]
```

For the dynamics of the twin rotor system, consider

$$
T_e = \begin{bmatrix} 0 \\ T_{e_2} \\ 0 \end{bmatrix},
\quad
\Omega = \begin{bmatrix} \Omega_1 \\ \Omega_2 \\ \Omega_3 \end{bmatrix},
\quad
\mathbb{I} =
\begin{bmatrix}
I_1 & 0 & 0 \\
0 & I_2 & 0 \\
0 & 0 & I_3
\end{bmatrix}
$$

Using the Euler rigid body equation,

$$
T_e + T_u = \Omega \times (\mathbb{I}\Omega) + \mathbb{I}\dot{\Omega}
$$

we obtain

$$
\begin{bmatrix} 0 \\ T_{e_2} \\ 0 \end{bmatrix}
+
\begin{bmatrix}
u_1 \\
0 \\
- u_2
\end{bmatrix}
=
\begin{bmatrix}
(I_3 - I_2)\Omega_2 \Omega_3 + I_1 \dot{\Omega}_1 \\
(I_1 - I_3)\Omega_1 \Omega_3 + I_2 \dot{\Omega}_2 \\
(I_2 - I_1)\Omega_1 \Omega_2 + I_3 \dot{\Omega}_3
\end{bmatrix}
$$

Recall that the angular velocity vector is given by

$$
\Omega =
\begin{bmatrix}
\dot{\phi} \\
\dot{\theta} \sin{\phi} \\
\dot{\theta} \cos{\phi}
\end{bmatrix}
$$

Differentiating with respect to time,

$$
\dot{\Omega} =
\begin{bmatrix}
\ddot{\phi} \\
\ddot{\theta} \sin{\phi} + \dot{\theta}\dot{\phi} \cos{\phi} \\
\ddot{\theta} \cos{\phi} - \dot{\theta}\dot{\phi} \sin{\phi}
\end{bmatrix}
$$

Substituting into the Euler equation yields

$$
u_1 = (I_3 - I_2)\,\dot{\theta}^2 \sin{\phi} \cos{\phi} + I_1 \ddot{\phi}
$$

$$
T_{e_2} = (I_1 - I_3)\,\dot{\phi}\dot{\theta} \cos{\phi}
+ I_2 \left(\ddot{\theta} \sin{\phi} + \dot{\theta}\dot{\phi} \cos{\phi}\right)
$$

$$
- u_2 = (I_2 - I_1)\,\dot{\phi}\dot{\theta} \sin{\phi}
+ I_3 \left(\ddot{\theta} \cos{\phi} - \dot{\theta}\dot{\phi} \sin{\phi}\right)
$$

Here, $T_{e_2}$ represents the reaction torque acting at the mechanical joint, which must be accounted for as a design constraint.

Therefore, the equations of motion governing the twin rotor system are

$$
\ddot{\phi} =
\frac{1}{I_1}
\left[
u_1 - (I_3 - I_2)\,\dot{\theta}^2 \sin{\phi} \cos{\phi}
\right]
$$

$$
\ddot{\theta} =
\frac{1}{\cos{\phi}}
\left[
\frac{1}{I_3}
\left(
(I_1 - I_2)\,\dot{\phi}\dot{\theta} \sin{\phi} - u_2
\right)
+ \dot{\theta}\dot{\phi} \sin{\phi}
\right]
$$

It can be observed that a mathematical singularity occurs when $\phi = 90^\circ$, known as gimbal lock. If a controller is designed based on this Euler angle formulation, the system becomes vulnerable to this issue, potentially leading to failure at this configuration. To avoid this, an alternative representation directly on $SO(3)$, without relying on $\phi$ and $\theta$, is introduced in the subsequent task using error dynamics.

```python
# ========================= System Dynamics ==================================
def rotor_system_dynamics(time, state_vector):
    theta, phi, theta_dot, phi_dot = state_vector

    tau1 = torque_u1(time)
    tau2 = torque_u2(time)

    # Avoid singularity when cos(phi) → 0 (gimbal lock condition)
    cos_phi = np.cos(phi)
    if abs(cos_phi) < 1e-5:
        cos_phi = 1e-5 * np.sign(cos_phi) if cos_phi != 0 else 1e-5

    # Equations of motion derived from Euler rigid body dynamics
    phi_ddot = (tau1 - (I_z - I_y) * (theta_dot**2) * np.sin(phi) * np.cos(phi)) / I_x
    theta_ddot = (-tau2 - (I_y - I_x - I_z) * theta_dot * phi_dot * np.sin(phi)) / (I_z * cos_phi)

    return [theta_dot, phi_dot, theta_ddot, phi_ddot]

# ============================ Simulation Setup ===============================
# Time interval and sampling points
time_range = (0, 10)
time_points = np.linspace(time_range[0], time_range[1], 300)

# Initial state: [theta, phi, theta_dot, phi_dot]
initial_state = [0.1, 0.0, 0.0, 0.0]

# Solve the differential equations
solution = solve_ivp(rotor_system_dynamics, time_range, initial_state,
                     t_eval=time_points, method='RK45')

theta_series = solution.y[0]
phi_series = solution.y[1]

# =================== Rotation & Geometry Construction =======================
def compute_rotation_matrix(theta, phi):
    Rz = np.array([
        [np.cos(theta), -np.sin(theta), 0],
        [np.sin(theta),  np.cos(theta), 0],
        [0,              0,             1]
    ])
    Rx = np.array([
        [1, 0,            0],
        [0, np.cos(phi), -np.sin(phi)],
        [0, np.sin(phi),  np.cos(phi)]
    ])
    return Rz @ Rx

# Generate animation frames for Plotly
animation_frames = []
support_length = 2.0   # Vertical support length
beam_length = 3.0      # Rotor beam length

for i in range(len(time_points)):
    theta_i = theta_series[i]
    phi_i = phi_series[i]
    rotation_matrix = compute_rotation_matrix(theta_i, phi_i)

    # 1. Vertical support (fixed base)
    support_coords = np.array([[0, 0, 0], [0, 0, -support_length]])

    # 2. Rotor beam (aligned with body axis)
    beam_local = np.array([[0, -beam_length/2, 0], [0, beam_length/2, 0]])
    beam_world = (rotation_matrix @ beam_local.T).T

    # 3. Rotor indicators at both ends
    rotor1_local = np.array([[0.5, -beam_length/2, 0], [-0.5, -beam_length/2, 0]])
    rotor2_local = np.array([[0, beam_length/2, 0.5], [0, beam_length/2, -0.5]])

    rotor1_world = (rotation_matrix @ rotor1_local.T).T
    rotor2_world = (rotation_matrix @ rotor2_local.T).T

    # Combine all line segments (using None to break segments)
    x_coords = [
        support_coords[0,0], support_coords[1,0], None,
        beam_world[0,0], beam_world[1,0], None,
        rotor1_world[0,0], rotor1_world[1,0], None,
        rotor2_world[0,0], rotor2_world[1,0]
    ]
    y_coords = [
        support_coords[0,1], support_coords[1,1], None,
        beam_world[0,1], beam_world[1,1], None,
        rotor1_world[0,1], rotor1_world[1,1], None,
        rotor2_world[0,1], rotor2_world[1,1]
    ]
    z_coords = [
        support_coords[0,2], support_coords[1,2], None,
        beam_world[0,2], beam_world[1,2], None,
        rotor1_world[0,2], rotor1_world[1,2], None,
        rotor2_world[0,2], rotor2_world[1,2]
    ]

    animation_frames.append(go.Frame(
        data=[go.Scatter3d(x=x_coords, y=y_coords, z=z_coords)]
    ))

# =========================== Visualization ==================================
fig = go.Figure(
    data=[go.Scatter3d(
        x=animation_frames[0].data[0].x,
        y=animation_frames[0].data[0].y,
        z=animation_frames[0].data[0].z,
        mode='lines',
        line=dict(color='blue', width=6)
    )],
    layout=go.Layout(
        title="Twin Rotor Motion Simulation",
        scene=dict(
            xaxis=dict(range=[-3, 3], title='X (Global)'),
            yaxis=dict(range=[-3, 3], title='Y (Global)'),
            zaxis=dict(range=[-3, 3], title='Z (Global)'),
            aspectmode='cube'
        ),
        updatemenus=[dict(
            type="buttons",
            buttons=[dict(
                label="Play",
                method="animate",
                args=[None, dict(frame=dict(duration=20, redraw=True), fromcurrent=True)]
            )]
        )]
    ),
    frames=animation_frames
)

fig.show()
```

**Task 3: Derivation and Simulation of a PID Controller for the Tracking Problem using Error Dynamics**

Let $(o_r, R_r) \in SE(3)$ denote the desired (reference) configuration of the twin rotor system. Since the centre of mass is fixed at the pivot point, we have $o_r \equiv o$. Therefore, the analysis can be restricted to the rotational dynamics described by $R$.

To achieve asymptotic tracking of a sufficiently smooth reference trajectory $R_r$, define the configuration error $R_e \in SO(3)$ as follows.

The left-invariant error (associated with the body frame) is given by

$$
R R_e = R_r
$$

$$
R_e = R^T R_r
$$

The right-invariant error (associated with the Earth frame) is defined as

$$
R_e R = R_r
$$

$$
R_e = R_r R^T
$$

Since the equations of motion are expressed in the Earth frame $\mathbf{e}$, the right-invariant error formulation is used for subsequent analysis.

\textbf{Part I: Derivation of Tracking Error Dynamics}

Let $\omega_r$ denote the spatial angular velocity corresponding to the desired trajectory, such that

$$
\widehat{\omega}_r = \dot{R}_r R_r^T
$$

To define the error in spatial angular velocity, consider

$$
\widehat{\omega}_e \triangleq \dot{R}_e R_e^T
$$

Substituting for $\dot{R}_e$,

$$
\widehat{\omega}_e = (\dot{R}_r R^T + R_r \dot{R}^T) R_e^T
$$

Rearranging terms,

$$
\widehat{\omega}_e = \dot{R}_r (R^T R_e^T) + (R_e R)\dot{R}^T R_e^T
$$

$$
\widehat{\omega}_e = \dot{R}_r R_r^T + R_e (R \dot{R}^T) R_e^T
$$

Using the identity $R R^T = I$, which implies

$$
R \dot{R}^T + \dot{R} R^T = 0
$$

we obtain

$$
\widehat{\omega}_e = \dot{R}_r R_r^T + R_e (-\dot{R} R^T) R_e^T
$$

$$
\widehat{\omega}_e = \widehat{\omega}_r - R_e \widehat{\omega} R_e^T
$$

$$
\widehat{\omega}_e = \widehat{\omega}_r - \widehat{R_e \omega}
$$

Hence, the spatial angular velocity error is given by

$$
\omega_e = \omega_r - R_e \omega
$$

For the error in spatial angular momentum, define the reference spatial angular momentum as

$$
\pi_r \triangleq R_r \mathbb{I} R_r^T \omega_r
$$

The error in spatial angular momentum is defined by

$$
\pi_e \triangleq R \mathbb{I} R_r^T \omega_e
$$

Substituting $\omega_e = \omega_r - R_e \omega$,

$$
\pi_e = R \mathbb{I} R_r^T (\omega_r - R_e \omega)
$$

$$
\pi_e = (R_e^T R_r)\mathbb{I} R_r^T \omega_r - R \mathbb{I} R^T \omega
$$

$$
\pi_e = R_e^T \pi_r - \pi
$$

Differentiating with respect to time,

$$
\dot{\pi}_e = \dot{R}_e^T \pi_r + R_e^T \dot{\pi}_r - \dot{\pi}
$$

Using $\dot{R}_e^T = R_e^T \widehat{\omega}_e^T$,

$$
\dot{\pi}_e = (R_e^T \widehat{\omega}_e^T)\pi_r + R_e^T \dot{\pi}_r - \dot{\pi}
$$

Since $\widehat{\omega}_e^T = -\widehat{\omega}_e$,

$$
\dot{\pi}_e = R_e^T (-\widehat{\omega}_e)\pi_r + R_e^T \dot{\pi}_r - \dot{\pi}
$$

$$
\dot{\pi}_e = R_e^T (\dot{\pi}_r - \widehat{\omega}_e \pi_r) - \dot{\pi}
$$

$$
\dot{\pi}_e = R_e^T (\dot{\pi}_r - \omega_e \times \pi_r) - \dot{\pi}
$$

Using $\pi_r = R_r \Pi_r$,

$$
\dot{\pi}_e = R_e^T (\dot{R}_r \Pi_r + R_r \dot{\Pi}_r - \omega_e \times \pi_r) - \dot{\pi}
$$

$$
\dot{\pi}_e = R_e^T (\widehat{\omega}_r R_r \Pi_r + R_r \dot{\Pi}_r - \omega_e \times \pi_r) - \dot{\pi}
$$

$$
\dot{\pi}_e = R_e^T (R_r \dot{\Pi}_r + \widehat{\omega}_r \pi_r - \omega_e \times \pi_r) - \dot{\pi}
$$

$$
\dot{\pi}_e = R_e^T (R_r \dot{\Pi}_r + (\omega_r - \omega_e)\times \pi_r) - \dot{\pi}
$$

$$
\dot{\pi}_e = R_e^T (R_r \dot{\Pi}_r + R_e \omega \times \pi_r) - \dot{\pi}
$$

$$
\dot{\pi}_e = R \dot{\Pi}_r + \omega \times (R_e^T \pi_r) - \dot{\pi}
$$

Hence, the resulting error dynamics of the system can be expressed as

$$
\dot{R}_e = \widehat{\omega}_e R_e
$$

$$
\dot{\pi}_e = \left( R \dot{\Pi}_r + \omega \times R_e^T \pi_r \right) - \tau_u - \tau_e
$$

**Part II: Stability Analysis**

Recall that

$$
R_e = R_r R^T \in SO(3)
$$

As $R \to R_r$, it follows that $R_e \to I$. To characterize this deviation, define a potential energy function $\mathcal{V}(R_e) \in \mathbb{R}_{\geq 0}$ such that $\mathcal{V}(R_e) \to 0$ at equilibrium when $R_e \to I$:

$$
\mathcal{V}(R_e) = \frac{1}{2} \text{trace}\big(K(I - R_e)\big)
$$

where $K$ is a positive definite weighting matrix determined by the inertia terms, given by

$$
K =
\begin{bmatrix}
\mathbb{I}_2 + \mathbb{I}_3 - \mathbb{I}_1 & 0 & 0 \\
0 & \mathbb{I}_1 + \mathbb{I}_3 - \mathbb{I}_2 & 0 \\
0 & 0 & \mathbb{I}_1 + \mathbb{I}_2 - \mathbb{I}_3
\end{bmatrix}
$$

To determine the rate of change of the potential energy, consider

$$
\mathcal{V}(R_e) = \frac{1}{2} \text{trace}(K(I - R_e))
$$

Differentiating with respect to time,

$$
\dot{\mathcal{V}}(R_e) = -\frac{1}{2} \text{trace}(K \dot{R}_e)
$$

Using the relation $\dot{R}_e = \widehat{\omega}_e R_e$,

$$
\dot{\mathcal{V}}(R_e) = -\frac{1}{2} \text{trace}(K \widehat{\omega}_e R_e)
$$

Applying the cyclic property of the trace operator,

$$
\dot{\mathcal{V}}(R_e) = -\frac{1}{2} \text{trace}(\widehat{\omega}_e R_e K)
$$

Taking the transpose inside the trace,

$$
\dot{\mathcal{V}}(R_e) = -\frac{1}{2} \text{trace}\left((\widehat{\omega}_e R_e K)^T\right)
$$

$$
\dot{\mathcal{V}}(R_e) = -\frac{1}{2} \text{trace}(K^T R_e^T \widehat{\omega}_e^T)
$$

Since $K$ is symmetric ($K = K^T$) and $\widehat{\omega}_e$ is skew-symmetric ($\widehat{\omega}_e^T = -\widehat{\omega}_e$),

$$
\dot{\mathcal{V}}(R_e) = \frac{1}{2} \text{trace}(K R_e^T \widehat{\omega}_e)
$$

Rewriting,

$$
\dot{\mathcal{V}}(R_e) = \frac{1}{2} \text{trace}(\widehat{\omega}_e K R_e^T)
$$

Using the identity $\text{trace}(\widehat{\omega}_e K R_e^T) = -\text{trace}(\widehat{\omega}_e R_e K)$,

$$
\dot{\mathcal{V}}(R_e) = -\frac{1}{4} \text{trace}\left(\widehat{\omega}_e (R_e K - K R_e^T)\right)
$$

Using the property $\text{trace}(\hat{x}\hat{y}) = -2 x^T y$ and noting that $(R_e K - K R_e^T) \in \mathfrak{so}(3)$,

$$
\dot{\mathcal{V}}(R_e) = \omega_e^T \underbrace{\frac{1}{2}(R_e K - K R_e^T)^{\vee}}_{\nabla \mathcal{V}}
$$

Thus,

$$
\dot{\mathcal{V}}(R_e) = \omega_e^T \nabla \mathcal{V}
$$

Here, $\nabla \mathcal{V} \in \mathbb{R}^3$ represents the gradient of the potential energy with respect to rotation, i.e., its spatial rate of change.

At equilibrium, $\dot{\mathcal{V}} = 0$ implies that $\nabla \mathcal{V} = 0$ for all $\omega_e \in \mathbb{R}^3$. This leads to the condition

$$
R_e K - K R_e^T = 0
$$

Since $K$ is a diagonal and positive definite matrix, the solutions correspond to matrices $R'_e$ that are orthogonal, diagonal, and satisfy $\det(R_e) = 1$, such that

$$
\dot{\mathcal{V}}(R'_e) = 0 \quad \forall \, \omega_e
$$

The set of matrices satisfying these conditions is

$$
R'_e =
\left\{
\begin{bmatrix}
1 & 0 & 0 \\
0 & 1 & 0 \\
0 & 0 & 1
\end{bmatrix},
\begin{bmatrix}
1 & 0 & 0 \\
0 & -1 & 0 \\
0 & 0 & -1
\end{bmatrix},
\begin{bmatrix}
-1 & 0 & 0 \\
0 & 1 & 0 \\
0 & 0 & -1
\end{bmatrix},
\begin{bmatrix}
-1 & 0 & 0 \\
0 & -1 & 0 \\
0 & 0 & 1
\end{bmatrix}
\right\}
$$

Evaluating $\mathcal{V}$ and $\dot{\mathcal{V}}$ at these configurations:

- For $R_e = \text{diag}(1,1,1)$, $\dot{\mathcal{V}} = 0$ and $\mathcal{V} = 0$, indicating a stable equilibrium (global minimum).

- For $R_e = \text{diag}(1,-1,-1)$, $\dot{\mathcal{V}} = 0$ and $\mathcal{V} = 2\mathbb{I}_1 \neq 0$, indicating an unstable equilibrium (local maximum).

- For $R_e = \text{diag}(-1,1,-1)$, $\dot{\mathcal{V}} = 0$ and $\mathcal{V} = 2\mathbb{I}_2 \neq 0$, indicating an unstable equilibrium (local maximum).

- For $R_e = \text{diag}(-1,-1,1)$, $\dot{\mathcal{V}} = 0$ and $\mathcal{V} = 2\mathbb{I}_3 \neq 0$, indicating an unstable equilibrium (local maximum).

From these results, it follows that $R_e = I$ corresponds to the globally stable equilibrium, while the remaining configurations represent unstable equilibria.

These unstable configurations are referred to as antipodal points, corresponding to a $180^\circ$ rotation about one of the principal axes relative to the reference orientation. Any small perturbation will drive the system away from these points toward the stable equilibrium.

Therefore, in the absence of perturbations ($\omega_r = 0$), the actuators of the twin rotor system do not generate torque at these configurations despite the presence of a positional error ($\mathcal{V} \neq 0$). However, in practice, such states are not maintained due to inevitable disturbances and noise in the system.

**Part III: Design of the PID Controller**

Recall the error dynamics equation:

$$
\dot{\pi}_e = \left(R \dot{\Pi}_r + \omega \times R_e^T \pi_r \right) - \tau_u - \tau_e
$$

Define the actuator input as

$$
\tau_u =
\underbrace{\left(R \dot{\Pi}_r + \omega \times R_e^T \pi_r \right)}_{\text{Feedforward}}
-
\underbrace{\left(k_P e_R + k_D e_{\omega} + k_I e_{I_R}\right)}_{\text{Feedback}}
$$

where $e_R, e_{\omega}, e_{I_R} \in \mathbb{R}^3$ denote the error signals to be defined, and $k_P, k_D, k_I \in \mathbb{M}_{3 \times 3}$ are gain matrices of the form

$$
k_i =
\begin{bmatrix}
k_{i_1} & 0 & 0 \\
0 & 0 & 0 \\
0 & 0 & k_{i_3}
\end{bmatrix},
\quad \text{for } i = P, I, D
$$

Here, $k_{i_1}$ and $k_{i_3}$ allow independent tuning of pitch and yaw dynamics, while the roll gain is set to zero since roll motion is constrained.

Substituting this control law into the error dynamics leads to cancellation of nonlinear effects such as gyroscopic coupling and inertia terms:

$$
\dot{\pi}_e = k_P e_R + k_D e_{\omega} + k_I e_{I_R} - \tau_e
$$

The reaction torque $\tau_e$ arising from holonomic constraints can be neglected as it is compensated by the mechanical structure. Therefore,

$$
\dot{\pi}_e = k_P e_R + k_D e_{\omega} + k_I e_{I_R}
$$

Thus, the resulting error dynamics become approximately linear, resembling a standard PID control structure in Euclidean space.

For the proportional error function $e_R$,

define it such that it grows with the gradient of the potential function (i.e., when $\dot{\mathcal{V}} < 0$), thereby driving the system toward the minimum potential $\mathcal{V} = 0$, which corresponds to the desired configuration:

$$
e_R = \nabla \mathcal{V} = \frac{1}{2}(K R_e^T - R_e K)^{\vee}
$$

---

For the derivative error function $e_{\omega}$,

as the angular velocity increases (i.e., when $\dot{\mathcal{K}} < 0$), the derivative error term increases, helping to reduce the velocity and ensuring that the twin rotor reaches the desired trajectory without oscillations or overshoot:

$$
e_{\omega} = \pi_e
$$

$$
e_{\omega} = R_e^T \pi_r - \pi
$$

---

For the integral error function $e_{I_R}$,

this term represents the accumulated orientation error over time, which helps eliminate steady-state errors caused by effects such as friction or persistent disturbances (e.g., airflow):

$$
e_{I_R} = \int e_R \, dt
$$

Thus, the control law can be expressed as

$$
\tau_u =
\left(R \dot{\Pi}_r + \omega \times R_e^T \pi_r \right)
- k_P e_R - k_D e_{\omega} - k_I e_{I_R}
$$

The feedforward component accounts for the required actuator torque to compensate for inertia and gyroscopic effects, while the feedback component corrects errors in orientation, angular velocity, and steady-state disturbances.

Recall that

$$
\tau_u = R T_u = R
\begin{bmatrix}
u_1 \\
0 \\
- u_2
\end{bmatrix}
$$

Hence,

$$
\begin{bmatrix}
u_1 \\
0 \\
- u_2
\end{bmatrix}
=
R^T
\left[
\left(R \dot{\Pi}_r + \omega \times R_e^T \pi_r \right)
- k_P e_R
- k_D (R_e^T \pi_r - \pi)
- k_I e_{I_R}
\right]
$$

Assuming equal gains along the active axes, i.e., $k_{i_1} = k_{i_3}$, the gain matrices reduce to scalar gains $k_P, k_D, k_I \in \mathbb{R}$. Therefore, the control input simplifies to

$$
\begin{bmatrix}
u_1 \\
0 \\
- u_2
\end{bmatrix}
=
\left(\dot{\Pi}_r + \Omega \times \Pi_r \right)
- k_P R^T e_R
- k_D (\Pi_r - \Pi)
- k_I R^T e_{I_R}
$$
