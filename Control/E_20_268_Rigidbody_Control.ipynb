# Task #1 — Dynamic Model of the Twin Rotor System

We derive the dynamic model of the twin rotor system and show that it can be written as

$$
\dot{R} = \hat{\omega} R,
$$

$$
\dot{\pi} = \tau^e + \tau^u,
$$

with

$$
\omega = (R I R^T)^{-1}\pi.
$$

---

# 1. Configuration of the System

The configuration space of the twin rotor is

$$
R \in SO(3).
$$

The orientation is parameterized using two rotations:

- Rotation about $e_3$ by angle $\theta$
- Rotation about the new $b_1$ axis by angle $\phi$

Thus,

$$
R = R_1(\phi) R_3(\theta),
$$

where

$$
R_1(\phi) =
\begin{bmatrix}
1 & 0 & 0 \\
0 & \cos\phi & -\sin\phi \\
0 & \sin\phi & \cos\phi
\end{bmatrix},
$$

$$
R_3(\theta) =
\begin{bmatrix}
\cos\theta & -\sin\theta & 0 \\
\sin\theta & \cos\theta & 0 \\
0 & 0 & 1
\end{bmatrix}.
$$

---

# 2. Angular Velocity

Using the composition of rotations,

$$
\dot{R} = R \hat{\Omega},
$$

where $\Omega$ is the body angular velocity.

From the kinematics of the chosen parameterization,

$$
\Omega =
\begin{bmatrix}
\dot{\phi}\cos\theta \\
-\dot{\phi}\sin\theta \\
\dot{\theta}
\end{bmatrix}.
$$

The spatial angular velocity is therefore

$$
\omega = R \Omega.
$$

The hat map is defined as

$$
\hat{\omega} =
\begin{bmatrix}
0 & -\omega_3 & \omega_2 \\
\omega_3 & 0 & -\omega_1 \\
-\omega_2 & \omega_1 & 0
\end{bmatrix}.
$$

Hence the attitude kinematics becomes

$$
\boxed{\dot{R} = \hat{\omega} R}.
$$

---

# 3. Angular Momentum

Let the inertia matrix in the body frame be

$$
I =
\begin{bmatrix}
I_1 & 0 & 0 \\
0 & I_2 & 0 \\
0 & 0 & I_3
\end{bmatrix}.
$$

The spatial angular momentum is

$$
\pi = R I \Omega.
$$

Since $\Omega = R^T \omega$, we obtain

$$
\pi = R I R^T \omega.
$$

Solving for $\omega$ gives

$$
\boxed{\omega = (R I R^T)^{-1}\pi}.
$$

---

# 4. Angular Momentum Dynamics

The rigid body rotational dynamics is

$$
\dot{\pi} = \tau,
$$

where the total torque is split into:

$$
\tau = \tau^e + \tau^u.
$$

Thus,

$$
\boxed{\dot{\pi} = \tau^e + \tau^u}.
$$

---

# 5. Control (Manipulable) Torque

The actuators generate torques in the $b_1$–$b_3$ plane.  
The control torque is

$$
\tau^u =
R
\begin{bmatrix}
u_1 \cos\alpha - u_2 \cos\beta \\
0 \\
u_1 \sin\alpha - u_2 \sin\beta
\end{bmatrix}.
$$

This represents the manipulable part of the torque.

---

# 6. Constraint Torque

The device is mechanically constrained about the $b_2$ axis.

Therefore the constraint torque takes the form

$$
\tau^e =
R
\begin{bmatrix}
0 \\
T_2 \\
0
\end{bmatrix}.
$$

To determine $T_2$, we use the body-frame Euler equation

$$
I \dot{\Omega} + \Omega \times (I\Omega) = T,
$$

and enforce zero motion along the constrained direction $e_2$.

Projecting along $e_2$ gives

$$
T_2
=
e_2^T
\left(
\Omega \times I\Omega + I\dot{\Omega}
\right).
$$

Thus,

$$
\boxed{
\tau^e =
R
\begin{bmatrix}
0 \\
e_2^T(\Omega \times I\Omega + I\dot{\Omega}) \\
0
\end{bmatrix}
}.
$$

---

# 7. Final Dynamic Model

Combining kinematics and dynamics, the twin rotor system is modeled by

$$
\boxed{\dot{R} = \hat{\omega} R}
$$

$$
\boxed{\dot{\pi} = \tau^e + \tau^u}
$$

$$
\boxed{\omega = (R I R^T)^{-1}\pi}
$$

with

$$
\tau^u =
R
\begin{bmatrix}
u_1 \cos\alpha - u_2 \cos\beta \\
0 \\
u_1 \sin\alpha - u_2 \sin\beta
\end{bmatrix}
$$

and

$$
\tau^e =
R
\begin{bmatrix}
0 \\
e_2^T(\Omega \times I\Omega + I\dot{\Omega}) \\
0
\end{bmatrix}.
$$

---

Therefore, the twin rotor system satisfies the required dynamic model.
