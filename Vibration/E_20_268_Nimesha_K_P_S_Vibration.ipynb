# Activity #1
Derive the 2-dof freedom model that will capture the first two dominant modes of small amplitude vibration of the system.

---


## 1. Governing Equation

For small-amplitude transverse vibrations of a slender beam, the Euler--Bernoulli beam theory gives the governing partial differential equation as

$$\rho A \frac{\partial^2 y(x,t)}{\partial t^2} + c \frac{\partial y(x,t)}{\partial t} + EI \frac{\partial^4 y(x,t)}{\partial x^4} + P \frac{\partial^2 y(x,t)}{\partial x^2} = f(x,t)$$

where $\rho$ is the material density, $A$ is the cross-sectional area, $E$ is Young's modulus, $I$ is the second moment of area, $c$ is the distributed viscous damping coefficient, $P$ is the axial compressive force, and $f(x,t)$ is the external transverse load per unit length.

For the present case, axial loading is absent $(P=0)$, hence the governing equation reduces to

$$\rho A \frac{\partial^2 y}{\partial t^2} + c \frac{\partial y}{\partial t} + EI \frac{\partial^4 y}{\partial x^4} = f(x,t)$$

---

## 2. Boundary Conditions

The beam is elastically restrained in rotation at the base and free at the tip. Accordingly, the boundary conditions are

$$
\begin{aligned}
y(0,t) &= 0 \\
EI \frac{\partial^2 y}{\partial x^2}(0,t) - k_\theta \frac{\partial y}{\partial x}(0,t) &= 0 \\
\frac{\partial^2 y}{\partial x^2}(L,t) &= 0 \\
\frac{\partial^3 y}{\partial x^3}(L,t) &= 0
\end{aligned}
$$

where the rotational stiffness at the base is defined as

$$k_\theta = k_{\text{spring}} \, l_{\text{spring}}^2$$



---

## 3. Operator Formulation

Let $\mathcal{F}$ denote the space of four-times continuously differentiable functions defined on $[0,L]$. Define the linear operator $H : \mathcal{F} \rightarrow \mathcal{F}$ as

$$H = \frac{EI}{\rho A} \frac{\partial^4}{\partial x^4}$$

The equation of motion can then be written in operator form as

$$\frac{\partial^2 y}{\partial t^2} + C \frac{\partial y}{\partial t} + H y = q(x,t)$$

where $c = C \rho A$ and $f(x,t) = \rho A q(x,t)$.

---

## 4. Free Vibration and Eigenvalue Problem

Neglecting damping and external forcing, the free vibration problem is governed by

$$H \psi_k(x) = \omega_k^2 \psi_k(x)$$

Substituting the operator definition gives

$$\frac{EI}{\rho A} \frac{d^4 \psi_k}{dx^4} = \omega_k^2 \psi_k$$

or equivalently,

$$\frac{d^4 \psi_k}{dx^4} = \beta_k^4 \psi_k, \qquad \beta_k^4 = \frac{\rho A \omega_k^2}{EI}$$

---

## 5. General Solution

The general solution of the fourth-order differential equation is

$$\psi_k(x) = A_k \cosh(\beta_k x) + B_k \sinh(\beta_k x) + C_k \cos(\beta_k x) + D_k \sin(\beta_k x)$$

---

## 6. Application of Boundary Conditions

### 6.1 Base End $(x=0)$

From the displacement boundary condition,

$$\psi_k(0) = 0 \quad \Rightarrow \quad A_k + C_k = 0$$

hence,

$$C_k = -A_k$$

The slope and curvature at the base are

$$
\begin{aligned}
\psi_k'(0) &= \beta_k (B_k + D_k) \\
\psi_k''(0) &= 2 \beta_k^2 A_k
\end{aligned}
$$

Substituting into the rotational spring condition yields

$$2 EI \beta_k^2 A_k - k_\theta \beta_k (B_k + D_k) = 0$$

### 6.2 Free End $(x=L)$

The moment-free condition gives

$$A_k \left[\cosh(\beta_k L) + \cos(\beta_k L)\right] + B_k \sinh(\beta_k L) - D_k \sin(\beta_k L) = 0$$

The shear-free condition gives

$$A_k \left[\sinh(\beta_k L) - \sin(\beta_k L)\right] + B_k \cosh(\beta_k L) - D_k \cos(\beta_k L) = 0$$

---

## 7. Characteristic Equation

The three boundary equations form a homogeneous linear system

$$\mathbf{H}(\beta_k) \begin{bmatrix} A_k \\ B_k \\ D_k \end{bmatrix} = \mathbf{0}$$

Non-trivial solutions exist only when

$$\det \mathbf{H}(\beta_k) = 0$$

Solving this characteristic equation yields the admissible values of $\beta_k$, from which the natural frequencies follow as

$$\omega_k = \beta_k^2 \sqrt{\frac{EI}{\rho A}}$$

---

## 8. Mode Shapes

The corresponding mode shapes are given by

$$\psi_k(x) = A_k \cosh(\beta_k x) + B_k \sinh(\beta_k x) - A_k \cos(\beta_k x) + D_k \sin(\beta_k x)$$

where the constants are determined up to an arbitrary scaling factor.



---

## 9. Forced Vibration Using Modal Expansion

Consider a harmonic point excitation applied at $x = l_{\text{exciter}}$:

$$q(x,t) = Q \sin(\Omega t)\, \delta(x - l_{\text{exciter}})$$

The displacement response is expanded in modal coordinates as

$$y(x,t) = \sum_{k=1}^{\infty} z_k(t)\, \psi_k(x)$$

Substituting into the governing equation and applying modal orthogonality leads to the decoupled equations

$$m_k \ddot z_k + 2 \zeta_k \omega_k m_k \dot z_k + m_k \omega_k^2 z_k = Q \psi_k(l_{\text{exciter}}) \sin(\Omega t)$$

where the modal mass is

$$m_k = \int_0^L \rho A \psi_k^2(x)\, dx$$

and

$$\zeta_k = \frac{C}{2 \omega_k}$$

---

## 10. Steady-State Solution

The steady-state solution of each modal coordinate is

$$z_k(t) = \chi_k(\Omega) \cos\!\left(\Omega t + \phi_k(\Omega)\right)$$

with amplitude

$$\chi_k(\Omega) = \left| \frac{Q \psi_k(l_{\text{exciter}})} {m_k \sqrt{(\omega_k^2 - \Omega^2)^2 + 4 \zeta_k^2 \omega_k^2 \Omega^2}} \right|$$

and phase lag

$$\phi_k(\Omega) = \tan^{-1} \left( \frac{2 \zeta_k \omega_k \Omega} {\omega_k^2 - \Omega^2} \right)$$

---

## 11. Modal Damping Matrix

The damping matrix for the beam modes is diagonalized through the orthogonality of the mode shapes. For the first two dominant modes, the beam damping matrix is:

$$\mathbf{C}_{\text{beam}} = \begin{bmatrix} 2 \zeta_1 \omega_1 m_1 & 0 \\ 0 & 2 \zeta_2 \omega_2 m_2 \end{bmatrix}$$

---

## 12. Reduction to a 2-DOF Model

Retaining only the first two dominant modes,

$$y(x,t) \approx z_1(t)\psi_1(x) + z_2(t)\psi_2(x)$$

The reduced equations of motion are

$$\mathbf{M}_{\text{beam}} \ddot{\mathbf z} + \mathbf{C}_{\text{beam}} \dot{\mathbf z} + \mathbf{K}_{\text{beam}} \mathbf z = \mathbf{F}_{\text{exciter}}$$

where

$$\mathbf{M}_{\text{beam}} = \begin{bmatrix} m_1 & 0 \\ 0 & m_2 \end{bmatrix}, \quad \mathbf{K}_{\text{beam}} = \begin{bmatrix} m_1 \omega_1^2 & 0 \\ 0 & m_2 \omega_2^2 \end{bmatrix}$$

---

## 13. Inclusion of Localised Attachments

Using Rayleigh's approximation, the effective mass for localised attachments at position $l_i$ is determined by the kinetic energy contribution. For the spring attachment:

$$\mathbf{M}_{\text{spring}} = m_{\text{eff,s}} \begin{bmatrix} \psi_1^2(l_{\text{spring}}) & \psi_1(l_{\text{spring}})\psi_2(l_{\text{spring}}) \\ \psi_1(l_{\text{spring}})\psi_2(l_{\text{spring}}) & \psi_2^2(l_{\text{spring}}) \end{bmatrix}$$

where $m_{\text{eff,s}} = \frac{1}{3} m_{\text{spring}} + m_{\text{fixed,s}}$.

Similarly, for the localised damper and exciter with masses $m_d$ and $m_e$ at positions $l_{\text{damper}}$ and $l_{\text{exciter}}$:

$$\mathbf{M}_{\text{damper}} = m_d \begin{bmatrix} \psi_1^2(l_{\text{damper}}) & \psi_1(l_{\text{damper}})\psi_2(l_{\text{damper}}) \\ \psi_1(l_{\text{damper}})\psi_2(l_{\text{damper}}) & \psi_2^2(l_{\text{damper}}) \end{bmatrix}$$

$$\mathbf{M}_{\text{exciter}} = m_e \begin{bmatrix} \psi_1^2(l_{\text{exciter}}) & \psi_1(l_{\text{exciter}})\psi_2(l_{\text{exciter}}) \\ \psi_1(l_{\text{exciter}})\psi_2(l_{\text{exciter}}) & \psi_2^2(l_{\text{exciter}}) \end{bmatrix}$$

---

## 14. Localized Stiffness and Damping Matrices

For a localized damper with coefficient $c_d$ and localized spring $k_d$, the additional matrices are:

$$\mathbf{C}_{\text{damper}} = c_d \begin{bmatrix} \psi_1^2(l_{\text{damper}}) & \psi_1(l_{\text{damper}})\psi_2(l_{\text{damper}}) \\ \psi_1(l_{\text{damper}})\psi_2(l_{\text{damper}}) & \psi_2^2(l_{\text{damper}}) \end{bmatrix}$$

$$\mathbf{K}_{\text{attachment}} = k_d \begin{bmatrix} \psi_1^2(l_{\text{spring}}) & \psi_1(l_{\text{spring}})\psi_2(l_{\text{spring}}) \\ \psi_1(l_{\text{spring}})\psi_2(l_{\text{spring}}) & \psi_2^2(l_{\text{spring}}) \end{bmatrix}$$

---

## 15. The Modal Force Vector

The vector $\mathbf{F}_{\text{exciter}}$ represents the projection of the external point force onto the chosen mode shapes. For a 2-DOF model, this is defined as:

$$\mathbf{F}_{\text{exciter}} = \begin{bmatrix} Q \psi_1(l_{\text{exciter}}) \\ Q \psi_2(l_{\text{exciter}}) \end{bmatrix} \sin(\Omega t)$$

---

## 16. Final Reduced System

The final 2-DOF system is

$$\left( \mathbf{M}_{\text{beam}} + \mathbf{M}_{\text{spring}} + \mathbf{M}_{\text{damper}} + \mathbf{M}_{\text{exciter}} \right) \ddot{\mathbf z} + \left( \mathbf{C}_{\text{beam}} + \mathbf{C}_{\text{damper}} \right) \dot{\mathbf z} + \left( \mathbf{K}_{\text{beam}} + \mathbf{K}_{\text{attachment}} \right) \mathbf z = \mathbf{F}_{\text{exciter}}$$

## Numerical Solution of $\det \mathbf{H}(\beta_k)=0$ and Mode Shape Determination



```python
# ===================== Imports =====================
import numpy as np
from numpy.linalg import det
from scipy.linalg import svd
from scipy.optimize import brentq

# ===================== Characteristic Determinant =====================
def det_characteristic(beta_val, L, EI_val, k_rot):
    """Determinant of characteristic matrix"""
    beta_L = beta_val * L

    H_mat = np.array([
        [2 * EI_val * beta_val**2, -k_rot * beta_val, -k_rot * beta_val],
        [np.cosh(beta_L) + np.cos(beta_L), np.sinh(beta_L), -np.sin(beta_L)],
        [np.sinh(beta_L) - np.sin(beta_L), np.cosh(beta_L), -np.cos(beta_L)]
    ])

    return det(H_mat)

# ===================== Root Finder =====================
def solve_betas(L, EI_val, k_rot, n_modes=2, beta_max=50):
    roots = []
    beta_scan = np.linspace(1e-4, beta_max, 5000)

    det_values = [det_characteristic(b, L, EI_val, k_rot) for b in beta_scan]

    for i in range(len(beta_scan) - 1):
        if det_values[i] * det_values[i + 1] < 0:
            beta_root = brentq(
                det_characteristic,
                beta_scan[i],
                beta_scan[i + 1],
                args=(L, EI_val, k_rot)
            )
            roots.append(beta_root)

            if len(roots) == n_modes:
                break

    return np.array(roots)

# ===================== Mode Shape Coefficients =====================
def compute_mode_coeffs(beta_val, L, EI_val, k_rot):
    beta_L = beta_val * L

    H_mat = np.array([
        [2 * EI_val * beta_val**2, -k_rot * beta_val, -k_rot * beta_val],
        [np.cosh(beta_L) + np.cos(beta_L), np.sinh(beta_L), -np.sin(beta_L)],
        [np.sinh(beta_L) - np.sin(beta_L), np.cosh(beta_L), -np.cos(beta_L)]
    ])

    # Null space via SVD
    _, _, Vh = svd(H_mat)
    null_vec = Vh[-1, :]     # [A, B, D]

    A_c, B_c, D_c = null_vec
    C_c = -A_c

    return A_c, B_c, C_c, D_c
```

## Final Reduced-Order Equation of Motion of the Beam–Attachment System


```python
# ===================== Imports =====================
import numpy as np
from scipy.integrate import quad

# ===================== Beam Parameters =====================
width = 0.025
thickness = 0.01
area = width * thickness

beam_mass = 1.65
beam_length = 0.815

E_modulus = 2e11
I_sec = width * thickness**3 / 12
density = beam_mass / (area * beam_length)

# ===================== Spring Parameters =====================
k_lin = 3800
x_spring = 0.75
spring_mass = 0.388
attached_mass = 0.009

k_theta = k_lin * x_spring**2

# ===================== Damper Parameters =====================
x_damper = 0.2
damper_mass = 0.53

# ===================== Exciter Parameters =====================
exciter_mass = 4.2
x_exciter = 0.4

# ===================== Solve for Betas =====================
betas = solve_betas(beam_length, E_modulus * I_sec, k_theta, n_modes=2)
beta_1, beta_2 = betas

print(f"beta1 = {beta_1:.4f}, beta2 = {beta_2:.4f}")

# ===================== Mode Shape Function =====================
def mode_shape(x, beta_val):
    A_c, B_c, C_c, D_c = compute_mode_coeffs(
        beta_val, beam_length, E_modulus * I_sec, k_theta
    )
    return (
        A_c * np.cosh(beta_val * x)
        + B_c * np.sinh(beta_val * x)
        + C_c * np.cos(beta_val * x)
        + D_c * np.sin(beta_val * x)
    )

# ===================== Mode Shape Values =====================
phi1_s = mode_shape(x_spring, beta_1)
phi2_s = mode_shape(x_spring, beta_2)

phi1_d = mode_shape(x_damper, beta_1)
phi2_d = mode_shape(x_damper, beta_2)

phi1_e = mode_shape(x_exciter, beta_1)
phi2_e = mode_shape(x_exciter, beta_2)

print(f"psi1(l_spring) = {phi1_s:.4f}, psi2(l_spring) = {phi2_s:.4f}")
print(f"psi1(l_damper) = {phi1_d:.4f}, psi2(l_damper) = {phi2_d:.4f}")
print(f"psi1(l_exciter) = {phi1_e:.4f}, psi2(l_exciter) = {phi2_e:.4f}")

# ===================== Modal Mass =====================
def modal_mass(beta_val):
    integrand = lambda x: density * area * mode_shape(x, beta_val)**2
    m_val, _ = quad(integrand, 0, beam_length)
    return m_val

m_1 = modal_mass(beta_1)
m_2 = modal_mass(beta_2)

# ===================== Natural Frequencies =====================
omega_1 = beta_1**2 * np.sqrt((E_modulus * I_sec) / (density * area))
omega_2 = beta_2**2 * np.sqrt((E_modulus * I_sec) / (density * area))

print(f"Modal mass m1 = {m_1:.4f} kg, m2 = {m_2:.4f} kg")
print(f"Natural frequencies: omega1 = {omega_1:.4f}, omega2 = {omega_2:.4f}")

# ===================== System Matrices =====================
M_beam = np.diag([m_1, m_2])
K_beam = np.diag([m_1 * omega_1**2, m_2 * omega_2**2])

m_eff_spring = (1/3) * spring_mass + attached_mass
M_spring = m_eff_spring * np.array([
    [phi1_s**2, phi1_s * phi2_s],
    [phi1_s * phi2_s, phi2_s**2]
])

M_damper = damper_mass * np.array([
    [phi1_d**2, phi1_d * phi2_d],
    [phi1_d * phi2_d, phi2_d**2]
])

M_exciter = exciter_mass * np.array([
    [phi1_e**2, phi1_e * phi2_e],
    [phi1_e * phi2_e, phi2_e**2]
])

# ===================== Matrix Printer =====================
def print_mat(title, mat):
    print(title)
    for r in mat:
        print(" ", ["{:.3f}".format(v) for v in r])
    print()

print_mat("M_beam", M_beam)
print_mat("M_spring", M_spring)
print_mat("M_damper", M_damper)
print_mat("M_exciter", M_exciter)
print_mat("K_beam", K_beam)
```

# Activity 2: Estimation of Damping Ratios of the First Two Dominant Modes

## 1. Theoretical Basis

To estimate the damping ratios $\zeta_1$ and $\zeta_2$, we analyze the discretized 2-DOF equation of motion derived in Activity 1:

$$
\mathbf{M}\ddot{\mathbf{z}} + \mathbf{C}\dot{\mathbf{z}} + \mathbf{K}\mathbf{z} = \mathbf{0}
$$

The system damping is predominantly determined by the discrete viscous damper located at $x_{damper} = 0.2$ m. To select the appropriate damping model, we examine the system's mode shapes.

## 2. Damping Model Selection

Using the theoretical model from Activity 1, we observe the mode shape amplitudes at the damper location:
* Mode 1: The damper is near a node ($|\psi_1| \approx 0$).
* Mode 2: The damper is near an antinode ($|\psi_2| \gg |\psi_1|$).

Since the damping is significantly more effective at higher frequencies, we assume Rayleigh Proportional Damping:

$$
\mathbf{C} = \alpha \mathbf{M} + \beta \mathbf{K}
$$

where $\alpha$ is the mass-proportional coefficient and $\beta$ is the stiffness-proportional coefficient.

## 3. Derivation of the Damping Ratio Formula

The modal damping ratio $\zeta_k$ for the $k^{th}$ mode is defined by projecting the damping matrix onto the mode shape vector $\phi_k$:

$$
\zeta_k = \frac{\phi_k^T \mathbf{C} \phi_k}{2\omega_k \phi_k^T \mathbf{M} \phi_k}
$$

Substitute the Rayleigh damping expression $\mathbf{C} = \alpha \mathbf{M} + \beta \mathbf{K}$ into the numerator:

$$
\phi_k^T \mathbf{C} \phi_k = \phi_k^T (\alpha \mathbf{M} + \beta \mathbf{K}) \phi_k
$$
$$
\phi_k^T \mathbf{C} \phi_k = \alpha (\phi_k^T \mathbf{M} \phi_k) + \beta (\phi_k^T \mathbf{K} \phi_k)
$$

Recall the eigenvalue property $\mathbf{K}\phi_k = \omega_k^2 \mathbf{M}\phi_k$. Multiplying by $\phi_k^T$:
$$
\phi_k^T \mathbf{K} \phi_k = \omega_k^2 (\phi_k^T \mathbf{M} \phi_k)
$$

Substituting this back into the damping equation:

$$
\phi_k^T \mathbf{C} \phi_k = (\alpha + \beta \omega_k^2) (\phi_k^T \mathbf{M} \phi_k)
$$

Now, substituting this numerator back into the definition of $\zeta_k$:

$$
\zeta_k = \frac{(\alpha + \beta \omega_k^2) (\phi_k^T \mathbf{M} \phi_k)}{2\omega_k (\phi_k^T \mathbf{M} \phi_k)}
$$

Simplifying the expression (the modal mass terms cancel out):

$$
\zeta_k = \frac{1}{2} \left( \frac{\alpha}{\omega_k} + \beta \omega_k \right)
$$

## 4. Estimation Formula for $\zeta_2$

As justified in Section 2, the damper has negligible effect on the low-frequency Mode 1 (near a node), so we assume the mass-proportional term is zero ($\alpha \approx 0$). This leaves Stiffness Proportional Damping:

$$
\zeta_k = \frac{\beta \omega_k}{2}
$$

We can now form a ratio between the second and first modes to eliminate $\beta$:

$$
\frac{\zeta_2}{\zeta_1} = \frac{\frac{\beta \omega_2}{2}}{\frac{\beta \omega_1}{2}} = \frac{\omega_2}{\omega_1}
$$

Rearranging for $\zeta_2$:

$$
\zeta_2 = \zeta_1 \left( \frac{\omega_2}{\omega_1} \right)
$$

## 5. Estimation Procedure

Since the exciter's maximum operating frequency is typically below the second natural frequency ($\omega_2$), the second damping peak cannot be measured experimentally. The following procedure is used to estimate $\zeta_2$:

1.  Measure $\zeta_1$: Obtain the frequency response curve experimentally. Use the Half-Power Bandwidth Method on the first resonance peak to calculate the experimental value of $\zeta_1$.
2.  Compute Natural Frequencies: Use the theoretical model derived in Activity 1 to calculate the natural frequencies $\omega_1$ and $\omega_2$.
3.  Calculate $\zeta_2$: Substitute the measured $\zeta_1$ and the calculated frequencies into the derived formula:

$$
\zeta_{2, estimated} = \zeta_{1, measured} \times \frac{\omega_2}{\omega_1}
$$

```python
# --------------------------------------------------
# Free vibration analysis of the reduced system
# --------------------------------------------------

from scipy.linalg import eigh # Import eigh

# Assemble total mass and stiffness matrices
M_total = M_beam + M_spring + M_damper + M_exciter
K_total = K_beam   # undamped, stiffness from beam only

# Solve generalized eigenvalue problem
eig_values, eig_vectors = eigh(K_total, M_total)

# Natural frequencies (rad/s)
omega_vals = np.sqrt(eig_values)

print("Natural frequencies including attachments:")
for idx, omega in enumerate(omega_vals, start=1):
    freq_hz = omega / (2 * np.pi)
    print(f"Mode {idx}: {omega:.3f} rad/s = {freq_hz:.3f} Hz")

# Modal matrix
Phi_mat = eig_vectors
print_mat("Modal Matrix", Phi_mat)
```

```python
# --------------------------------------------------
# Beam-only mode shape plotting
# --------------------------------------------------

# Spatial discretisation
x_axis = np.linspace(0, beam_length, 1000)

# Mode shape evaluation
mode1 = mode_shape(x_axis, beta_1)
mode2 = mode_shape(x_axis, beta_2)

# Normalisation
mode1_n = mode1 / np.max(np.abs(mode1))
mode2_n = mode2 / np.max(np.abs(mode2))

# Plotting
import matplotlib.pyplot as plt
plt.figure(figsize=(7, 3))

plt.plot(x_axis, mode1_n, color='tab:green', linewidth=2,
         label=r'$\psi_1$ (First mode)')
plt.plot(x_axis, mode2_n, color='tab:red', linewidth=2,
         label=r'$\psi_2$ (Second mode)')

plt.xlabel('Beam length coordinate (m)')
plt.ylabel('Normalised mode shape amplitude')
plt.grid(True)
plt.legend()
plt.show()
```

$$



## Activity 3: Graphical Illustration of Selected System Responses

### Unforced Response Due to an Initial Displacement at the Spring End

The displacement vector of the complete system can be expressed using modal coordinates as

$$
\mathbf{z}(t) = \mathbf{\Phi}\,\mathbf{q}(t)
$$

$$
\begin{bmatrix}
z_1(t) \\
z_2(t)
\end{bmatrix}
=
\begin{bmatrix}
\phi_{11} & \phi_{12} \\
\phi_{21} & \phi_{22}
\end{bmatrix}
\begin{bmatrix}
q_1(t) \\
q_2(t)
\end{bmatrix}.
$$

For the beam alone, the truncated two-degree-of-freedom approximation of the transverse displacement is given by

$$
y(x,t) = z_1(t)\,\psi_1(x) + z_2(t)\,\psi_2(x)
$$

Substituting the modal transformation and regrouping terms yields

$$
\begin{aligned}
y(x,t) &=
(\phi_{11} q_1(t) + \phi_{12} q_2(t))\,\psi_1(x)
+ (\phi_{21} q_1(t) + \phi_{22} q_2(t))\,\psi_2(x) \\
&=
\underbrace{\left(\phi_{11}\psi_1(x) + \phi_{21}\psi_2(x)\right)}_{\psi_1^{*}(x)} q_1(t)
+
\underbrace{\left(\phi_{12}\psi_1(x) + \phi_{22}\psi_2(x)\right)}_{\psi_2^{*}(x)} q_2(t)
\end{aligned}
$$

Assuming an initial transverse displacement $y_0$ at the free end of the beam,

$$
y(l_{\text{beam}},0) = y_0
$$

which leads to

$$
y_0 = \psi_1^{*}(l_{\text{beam}})\,q_1(0)
+ \psi_2^{*}(l_{\text{beam}})\,q_2(0)
$$

Since the displacement is applied at the free end where the first mode has its maximum amplitude, the response is assumed to be dominated by the first mode. Accordingly,

$$
q_2(0) = 0,
\qquad
q_1(0) = \frac{y_0}{\psi_1^{*}(l_{\text{beam}})}
$$

The unforced modal equations of motion are given by

$$
\mathbf{M}_m \ddot{\mathbf{q}}
+ \mathbf{C}_m \dot{\mathbf{q}}
+ \mathbf{K}_m \mathbf{q} = 0
$$

Under the assumption of proportional damping, this can be written as

$$
\mathbf{M}_m \ddot{\mathbf{q}}
+ \beta\,\mathbf{K}_m \dot{\mathbf{q}}
+ \mathbf{K}_m \mathbf{q} = 0
$$

with the initial conditions

$$
\mathbf{q}(0) =
\begin{bmatrix}
q_1(0) \\ 0
\end{bmatrix},
\qquad
\dot{\mathbf{q}}(0) =
\ddot{\mathbf{q}}(0) =
\begin{bmatrix}
0 \\ 0
\end{bmatrix}
$$

where the damping parameter is defined as

$$
\beta = \frac{2\zeta_1}{\omega_1} = \frac{2\zeta_2}{\omega_2}
$$

Once the modal responses are obtained, the final expression for the free damped vibration of the beam is

$$
y(x,t) = \psi_1^{*}(x)\,q_1(t) + \psi_2^{*}(x)\,q_2(t)
$$


```python
# --------------------------------------------------
# Initial displacement at free end
# --------------------------------------------------
y_init = 0.003     # Initial displacement magnitude (m)

# Extract modal matrix components
phi_11, phi_12 = Phi_mat[0, 0], Phi_mat[0, 1]
phi_21, phi_22 = Phi_mat[1, 0], Phi_mat[1, 1]

# --------------------------------------------------
# Modified mode shapes including attachments
# --------------------------------------------------
def psi_star_mode1(x_val):
    return (
        phi_11 * mode_shape(x_val, beta_1)
        + phi_21 * mode_shape(x_val, beta_2)
    )

def psi_star_mode2(x_val):
    return (
        phi_12 * mode_shape(x_val, beta_1)
        + phi_22 * mode_shape(x_val, beta_2)
    )

# Evaluate modified first mode at free end
psi_star_L = psi_star_mode1(beam_length)
print(f"psi*_1(l_beam) = {psi_star_L:.4f}")

# Initial modal coordinate
q1_init = y_init / psi_star_L
print(f"Initial modal displacement q1(0) = {q1_init:.4f}")

# --------------------------------------------------
# Modal mass, stiffness, and damping matrices
# --------------------------------------------------
M_modal = Phi_mat.T @ M_total @ Phi_mat
K_modal = Phi_mat.T @ K_total @ Phi_mat

zeta_1 = 0.008
zeta_2 = 0.05

beta_damp = 2 * zeta_1 / omega_vals[0]
C_modal = beta_damp * K_modal
```

```python
# --------------------------------------------------
# State-space ODE for free vibration response
# --------------------------------------------------
from scipy.integrate import odeint # Import odeint

def modal_ode(state, time, M_inv, C_m, K_m):
    q1, q2, q1_dot, q2_dot = state

    q_vec = np.array([q1, q2])
    qd_vec = np.array([q1_dot, q2_dot])

    qdd_vec = M_inv @ (-C_m @ qd_vec - K_m @ q_vec)

    return [q1_dot, q2_dot, qdd_vec[0], qdd_vec[1]]

# Inverse of modal mass matrix
M_modal_inv = np.linalg.inv(M_modal)

# Initial conditions
state_0 = [q1_init, 0.0, 0.0, 0.0]

# Time vector
time_vec = np.arange(0, 10, 0.001)

# ODE integration
response = odeint(
    modal_ode,
    state_0,
    time_vec,
    args=(M_modal_inv, C_modal, K_modal)
)

# Modal responses
q1_t = response[:, 0]
q2_t = response[:, 1]

# --------------------------------------------------
# Physical displacement at free end
# --------------------------------------------------
z_t = np.zeros((len(time_vec), 2))

for i in range(len(time_vec)):
    z_t[i, :] = Phi_mat @ np.array([q1_t[i], q2_t[i]])

psi1_L = mode_shape(beam_length, beta_1)
psi2_L = mode_shape(beam_length, beta_2)

y_L_t = z_t[:, 0] * psi1_L + z_t[:, 1] * psi2_L

# Plot response
plt.figure(figsize=(10, 3))
plt.plot(time_vec, y_L_t)
plt.xlabel('Time (s)')
plt.ylabel('Displacement (m)')
plt.title('Free-End Unforced Response of the Beam')
plt.grid(True)
plt.show()
```

```python
# --------------------------------------------------
# Spatial–temporal displacement field
# --------------------------------------------------
y_xt = np.zeros((len(time_vec), len(x_axis)))

def psi_star1_full(x_vals):
    return (
        phi_11 * mode_shape(x_vals, beta_1)
        + phi_21 * mode_shape(x_vals, beta_2)
    )

def psi_star2_full(x_vals):
    return (
        phi_12 * mode_shape(x_vals, beta_1)
        + phi_22 * mode_shape(x_vals, beta_2)
    )

for i in range(len(time_vec)):
    y_xt[i, :] = (
        psi_star1_full(x_axis) * q1_t[i]
        + psi_star2_full(x_axis) * q2_t[i]
    )

# --------------------------------------------------
# Animation setup
# --------------------------------------------------
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

fig, ax = plt.subplots(figsize=(7, 3))
ax.set_xlim(0, beam_length)

y_lim = np.max(np.abs(y_xt)) * 1.1
ax.set_ylim(-y_lim, y_lim)

ax.set_xlabel('Beam axis (m)')
ax.set_ylabel('Transverse displacement (m)')
ax.set_title('Unforced Beam Vibration Response')
ax.grid(True)

beam_line, = ax.plot([], [], lw=2)

def init_anim():
    beam_line.set_data([], [])
    return beam_line,

def update_anim(i):
    beam_line.set_data(x_axis, y_xt[i, :])
    return beam_line,

frame_step = 10
frame_ids = range(0, len(time_vec), frame_step)

anim = FuncAnimation(
    fig,
    update_anim,
    init_func=init_anim,
    frames=frame_ids,
    interval=int((time_vec[1] - time_vec[0]) * frame_step * 1000),
    blit=True
)

plt.close(fig)
HTML(anim.to_html5_video())
```

### 3.2 Forced Vibration Response for Selected Motor Speeds

The forced vibration behaviour of the complete reduced-order system is examined for three different rotational speeds of the motor. The governing equations of motion in modal coordinates are expressed as

$$
\mathbf{M}_m \ddot{\mathbf{q}} + \mathbf{C}_m \dot{\mathbf{q}} + \mathbf{K}_m \mathbf{q} = \mathbf{F}_m .
$$

The modal forcing vector $\mathbf{F}_m$ is obtained by projecting the physical excitation force onto the modal subspace, such that

$$
\mathbf{F}_m = \mathbf{\Phi}^T \mathbf{F}_{\text{exciter}},
$$

where the excitation force generated by the rotating motor unbalance is given by

$$
\mathbf{F}_{\text{exciter}} =
\begin{bmatrix}
Q(\Omega)\,\psi_1(l_{\text{exciter}})\,\sin(\Omega t) \\[1mm]
Q(\Omega)\,\psi_2(l_{\text{exciter}})\,\sin(\Omega t)
\end{bmatrix}.
$$

The amplitude of the harmonic excitation due to the rotating eccentric mass is defined as

$$
Q(\Omega) = m_{\text{ecc}}\, r\, \Omega^2,
$$

where $m_{\text{ecc}}$ denotes the eccentric mass and $r$ represents the radial offset of the mass from the axis of rotation of the motor.

The resulting non-homogeneous system of ordinary differential equations is solved to obtain the particular modal responses $\mathbf{q}(t)$ associated with the external forcing. These modal responses are subsequently substituted into the physical displacement expression to evaluate the system response in space and time, given by

$$
y(x,t) = \psi_1^{*}(x)\,q_1(t) + \psi_2^{*}(x)\,q_2(t).
$$


```python
# --------------------------------------------------
# Motor excitation settings
# --------------------------------------------------

Omega_a = 32.472     # Close to first natural frequency
Omega_b = 100.0      # Mid-range excitation
Omega_c = 204.048    # Close to second natural frequency

Omega_set = [Omega_a, Omega_b, Omega_c]

# Eccentric mass properties
m_unb = 0.0045       # Unbalance mass (kg)
ecc_r = 0.035        # Eccentricity radius (m)

def forcing_amplitude(m_unb, ecc_r, Omega):
    """
    Computes harmonic force amplitude due to rotating unbalance
    """
    return 2.0 * m_unb * ecc_r * Omega**2
```

```python
# --------------------------------------------------
# Time domain for forced response
# --------------------------------------------------
time_f = np.arange(0, 7.5, 0.001)

def forced_modal_ode(state, time, M_inv, C_m, K_m, Omega, Q_amp, psi_exc):
    q1, q2, q1d, q2d = state

    q_vec = np.array([q1, q2])
    qd_vec = np.array([q1d, q2d])

    # Modal forcing vector
    f_modal = psi_exc * Q_amp * np.sin(Omega * time)

    # Second-order system
    qdd_vec = M_inv @ (f_modal - C_m @ qd_vec - K_m @ q_vec)

    return [q1d, q2d, qdd_vec[0], qdd_vec[1]]

# Initial state (rest)
state_init = [0.0, 0.0, 0.0, 0.0]

# Modified mode shapes evaluated at exciter location
psi_exciter = np.array([
    psi_star_mode1(x_exciter),
    psi_star_mode2(x_exciter)
])

# --------------------------------------------------
# Solve system for each excitation frequency
# --------------------------------------------------
forced_solutions = []

for Omega_val in Omega_set:

    Q_val = forcing_amplitude(m_unb, ecc_r, Omega_val)

    sol = odeint(
        forced_modal_ode,
        state_init,
        time_f,
        args=(M_modal_inv, C_modal, K_modal, Omega_val, Q_val, psi_exciter)
    )

    forced_solutions.append(sol)
```

```python
# --------------------------------------------------
# Animation of forced beam vibration
# --------------------------------------------------
for k, Omega_val in enumerate(Omega_set):

    sol = forced_solutions[k]
    q1_forced = sol[:, 0]
    q2_forced = sol[:, 1]

    # Space–time displacement matrix
    y_xt_forced = np.zeros((len(time_f), len(x_axis)))

    for i in range(len(time_f)):
        y_xt_forced[i, :] = (
            psi_star1_full(x_axis) * q1_forced[i]
            + psi_star2_full(x_axis) * q2_forced[i]
        )

    fig, ax = plt.subplots(figsize=(7, 3))
    ax.set_xlim(0, beam_length)

    disp_lim = np.max(np.abs(y_xt_forced)) * 1.1
    ax.set_ylim(-disp_lim, disp_lim)

    ax.set_xlabel('Beam coordinate (m)')
    ax.set_ylabel('Displacement (m)')
    ax.set_title(rf'Forced Beam Response, $\Omega = {Omega_val:.2f}$ rad/s')
    ax.grid(True)

    beam_line, = ax.plot([], [], lw=2)

    def init_anim():
        beam_line.set_data([], [])
        return beam_line,

    def update_anim(i):
        beam_line.set_data(x_axis, y_xt_forced[i, :])
        return beam_line,

    step = 10
    frame_ids = range(0, len(time_f), step)

    anim = FuncAnimation(
        fig,
        update_anim,
        init_func=init_anim,
        frames=frame_ids,
        interval=int((time_f[1] - time_f[0]) * step * 1000),
        blit=True
    )

    plt.close(fig)
    display(HTML(anim.to_html5_video()))
```

### 3.3 Analytical Frequency Response of the System Including the First Two Dominant Modes

The steady-state frequency response of the system is defined as the ratio between the displacement amplitude and the applied harmonic force amplitude. Accordingly, the frequency response function may be written as

$$
G(x,\Omega) = \frac{Y(x,\Omega)}{Q(\Omega)},
$$

where $Y(x,\Omega)$ denotes the displacement response in the frequency domain and $Q(\Omega)$ represents the excitation force amplitude.

To account for the influence of the attached components, the classical modal amplitude and phase expressions are reformulated for the complete system. The modified modal amplitude associated with the $k$th mode is expressed as

$$
\chi_k^{*}(\Omega)
=
\left|
\frac{Q(\Omega)\,\psi_k^{*}(l_{\text{exciter}})}
{M_{m_k}\sqrt{(\omega_k^2-\Omega^2)^2 + 4\zeta_k^2\Omega^2\omega_k^2}}
\right|,
$$

while the corresponding phase angle is given by

$$
\phi_k^{*}(\Omega)
=
\arctan\!\left(
\frac{2\zeta_k\Omega\omega_k}{\omega_k^2-\Omega^2}
\right).
$$

\textbf{Note:} The quantities $\omega_k$ and $\zeta_k$ denote the natural frequencies and damping ratios of the complete beam–attachment system.

Using the modified mode shapes, the physical displacement response can be written as a superposition of modal contributions,

\begin{align*}
y(x,t) &= \psi_1^{*}(x)\,q_1(t) + \psi_2^{*}(x)\,q_2(t) \\
&=
\chi_1^{*}(\Omega)\,\psi_1^{*}(x)\cos\!\bigl(\Omega t + \phi_1^{*}(\Omega)\bigr)
+
\chi_2^{*}(\Omega)\,\psi_2^{*}(x)\cos\!\bigl(\Omega t + \phi_2^{*}(\Omega)\bigr).
\end{align*}

Recalling that
$
\cos(\Omega t + \phi_k^{*}) = \Re\{e^{j(\Omega t + \phi_k^{*})}\},
$
the displacement may be rewritten in complex exponential form as

\begin{align*}
y(x,t)
&=
\Re\Bigl\{
\chi_1^{*}(\Omega)\,\psi_1^{*}(x)\,e^{j(\Omega t + \phi_1^{*}(\Omega))}
+
\chi_2^{*}(\Omega)\,\psi_2^{*}(x)\,e^{j(\Omega t + \phi_2^{*}(\Omega))}
\Bigr\} \\
&=
\Re\Bigl\{
\bigl[
\chi_1^{*}(\Omega)\,\psi_1^{*}(x)\,e^{j\phi_1^{*}(\Omega)}
+
\chi_2^{*}(\Omega)\,\psi_2^{*}(x)\,e^{j\phi_2^{*}(\Omega)}
\bigr]
e^{j\Omega t}
\Bigr\}.
\end{align*}

The term multiplying the harmonic factor $e^{j\Omega t}$ represents the spatial frequency-domain displacement, which is therefore defined as

$$
Y(x,\Omega)
=
\chi_1^{*}(\Omega)\,\psi_1^{*}(x)\,e^{j\phi_1^{*}(\Omega)}
+
\chi_2^{*}(\Omega)\,\psi_2^{*}(x)\,e^{j\phi_2^{*}(\Omega)}.
$$

Although $\chi_k^{*}(\Omega)$ and $\phi_k^{*}(\Omega)$ separately describe the magnitude and phase information, the frequency response formulation requires the full complex representation. Thus, reverting to the complex form yields

$$
Y(x,\Omega)
=
\frac{Q(\Omega)\,\psi_1^{*}(l_{\text{exciter}})}
{M_{m_1}\left[(\omega_1^2-\Omega^2)+j(2\zeta_1\Omega\omega_1)\right]}\,\psi_1^{*}(x)
+
\frac{Q(\Omega)\,\psi_2^{*}(l_{\text{exciter}})}
{M_{m_2}\left[(\omega_2^2-\Omega^2)+j(2\zeta_2\Omega\omega_2)\right]}\,\psi_2^{*}(x).
$$

Dividing by the excitation amplitude $Q(\Omega)$ leads to the final expression for the frequency response function,

$$
G(x,\Omega)
=
\frac{\psi_1^{*}(l_{\text{exciter}})\,\psi_1^{*}(x)}
{M_{m_1}\left[(\omega_1^2-\Omega^2)+j(2\zeta_1\Omega\omega_1)\right]}
+
\frac{\psi_2^{*}(l_{\text{exciter}})\,\psi_2^{*}(x)}
{M_{m_2}\left[(\omega_2^2-\Omega^2)+j(2\zeta_2\Omega\omega_2)\right]}.
$$

From this formulation, the magnitude response is obtained as $|G(x,\Omega)|$, while the phase response is given by $\arg\!\left(G(x,\Omega)\right)$.


```python
# --------------------------------------------------
# Beam locations selected for FRF evaluation
# --------------------------------------------------
x_loc_A = beam_length
x_loc_B = x_damper
x_loc_C = 0.495      # Close to second-mode node

x_probe = np.array([x_loc_A, x_loc_B, x_loc_C])

# Modified mode shapes at selected locations
phi1_x = np.array([psi_star_mode1(xi) for xi in x_probe])
phi2_x = np.array([psi_star_mode2(xi) for xi in x_probe])

# Modified mode shapes at exciter location
phi1_exc = psi_star_mode1(x_exciter)
phi2_exc = psi_star_mode2(x_exciter)

# Modal parameters
wn_1, wn_2 = omega_vals[0], omega_vals[1]
Mm_1 = M_modal[0, 0]
Mm_2 = M_modal[1, 1]

# --------------------------------------------------
# Frequency Response Function definition
# --------------------------------------------------
def FRF_at_point(idx, Omega):
    """
    Returns complex FRF at selected spatial location
    """
    D1 = (wn_1**2 - Omega**2) + 1j * (2 * zeta_1 * Omega * wn_1)
    D2 = (wn_2**2 - Omega**2) + 1j * (2 * zeta_2 * Omega * wn_2)

    G1 = (phi1_exc * phi1_x[idx]) / (Mm_1 * D1)
    G2 = (phi2_exc * phi2_x[idx]) / (Mm_2 * D2)

    return G1 + G2
```

```python
# --------------------------------------------------
# Frequency range for FRF computation
# --------------------------------------------------
Omega_range = np.linspace(0.01, 300, 600)

FRF_vals = np.zeros((3, len(Omega_range)), dtype=complex)

# Compute FRFs
for p in range(3):
    for k, Om in enumerate(Omega_range):
        FRF_vals[p, k] = FRF_at_point(p, Om)

# --------------------------------------------------
# Plotting: Magnitude and Phase
# --------------------------------------------------
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

colors_mag = ['tab:blue', 'tab:green', 'tab:red']
colors_phase = ['tab:purple', 'tab:orange', 'tab:brown']

# --- Magnitude plot (dB)
for p in range(3):
    axes[0].plot(
        Omega_range,
        20 * np.log10(np.abs(FRF_vals[p])),
        color=colors_mag[p],
        linewidth=2,
        label=rf"$x_{p+1}$"
    )

axes[0].set_xlabel(r"Excitation frequency $\Omega$ (rad/s)")
axes[0].set_ylabel(r"$20\log_{10}|G(x,\Omega)|$ [dB]")
axes[0].set_title("Frequency Response Magnitude")
axes[0].legend()
axes[0].grid(True)

# --- Phase plot
for p in range(3):
    axes[1].plot(
        Omega_range,
        np.unwrap(np.angle(FRF_vals[p])),
        color=colors_phase[p],
        linewidth=2,
        label=rf"$x_{p+1}$"
    )

axes[1].set_xlabel(r"Excitation frequency $\Omega$ (rad/s)")
axes[1].set_ylabel(r"Phase of $G(x,\Omega)$ [rad]")
axes[1].set_title("Frequency Response Phase")
axes[1].legend()
axes[1].grid(True)

plt.tight_layout()
plt.show()
```

### Activity 5: Design of a Tuned Mass Absorber for the First Mode

A tuned mass vibration absorber is designed to suppress vibrations near the first natural frequency.
Its effectiveness is evaluated using the free vibration response to an initial displacement applied
at the free end of the beam.


### Forced Response Near the First Mode with the Absorber Attached

image.png

From the preceding response plot, the first experimentally identified natural mode is observed at

$$
f_1 = 6.42~\text{Hz}.
$$

A schematic representation of the vibration absorber, designed for a rigid beam–spring system, is
considered next. Owing to the asymmetric motion about the plane containing the exciter motor, only
one of the two absorber beams is employed for vibration suppression. This effectively increases the
system order from one degree of freedom to two degrees of freedom.

The absorber is modelled as a cantilever beam with a concentrated mass $m_2$ attached at a distance
$l_{\text{absorber}}$ from its fixed end, which is located at $l_{\text{exciter}}$. According to
Euler–Bernoulli beam theory, the equivalent stiffness of the absorber beam is given by

$$
k_{\text{absorber}} = \frac{3 E I_{\text{absorber}}}{l_{\text{absorber}}^3}.
$$

Assuming that the absorber primarily responds through its first bending mode within the operating
frequency range, the effective modal mass is estimated using Rayleigh’s method as

$$
m_2 = m_{\text{mass}} + \frac{33}{140}\,m_{\text{absorber beam}}.
$$

The natural frequency of the absorber system can therefore be expressed as

$$
f = \frac{1}{2\pi}\sqrt{\frac{k}{m}}
= \frac{1}{2\pi}\sqrt{\frac{3 E I_{\text{absorber}}}{m_2\,l_{\text{absorber}}^3}}.
$$

Rearranging the above expression to determine the required location of the attached mass yields

$$
l_{\text{absorber}} =
\left[
\frac{3 E I_{\text{absorber}}}{m_2 (2\pi f)^2}
\right]^{\frac{1}{3}}.
$$

To ensure effective vibration suppression, the absorber is tuned such that its natural frequency
lies close to the first natural frequency of the primary system.


```python
# --------------------------------------------------
# Mechanical properties of the vibration absorber
# --------------------------------------------------

E_abs = 2.0e11             # Young's modulus of absorber material (Pa)
I_abs = 7.2179e-12         # Second moment of area (m^4)

m_total_abs = 0.438        # Total absorber assembly mass (kg)
m_slider = 0.121           # Tunable lumped mass (kg)
m_beam_abs = 0.126         # Absorber beam mass per unit length (kg/m)

# --------------------------------------------------
# Effective modal mass (Rayleigh approximation)
# --------------------------------------------------
beam_length_abs = 0.27     # Length of absorber beam (m)

m_eff_abs = m_slider + (33.0 / 140.0) * m_beam_abs * beam_length_abs
```

```python
# --------------------------------------------------
# Function to evaluate absorber mass position
# based on target tuning frequency
# --------------------------------------------------
import numpy as np # Import numpy

def absorber_mass_position(freq_hz):
    omega_target = 2.0 * np.pi * freq_hz

    stiffness_term = 3.0 * E_abs * I_abs
    inertia_term = m_eff_abs * omega_target**2

    l_abs = (stiffness_term / inertia_term)**(1.0 / 3.0)
    return l_abs


# Target frequency: first experimental natural mode
f_target = 6.42

l_abs_value = absorber_mass_position(f_target)

print(f"Calculated absorber mass position: {l_abs_value:.4f} m")
```

Accordingly, the required distance of the movable mass from the absorber base,
tuned to mitigate vibrations associated with the first mode, is obtained as

$$
l_{\text{absorber}} = 0.2743~\text{m}.
$$


### Simulation-Based Justification

#### Approach 1: Elastic Beam Model (Euler–Bernoulli Theory)

To re-evaluate the dynamic response of the beam–absorber system, a simplified
single-degree-of-freedom (1-DOF) representation is adopted. Since the vibration
response is dominated by the first mode, the displacement field can be reduced as

$$
y(x,t) = z_1(t)\psi_1(x) + z_2(t)\psi_2(x)
\;\;\Rightarrow\;\;
y(x,t) = z_1(t)\psi_1(x).
$$

With this approximation, the equation of motion for the complete assembly,
including the absorber, is written as

$$
m_a \ddot{z} + c_a \dot{z} + k_a z
= Q(\Omega)\,\psi_1(l_{\text{exciter}})\,\sin(\Omega t).
$$

The equivalent modal mass of the system is given by

$$
m_a =
m_{m_1}
+ m_{\text{eff}}\,\psi_1^2(l_{\text{spring}})
+ m_{\text{damper}}\,\psi_1^2(l_{\text{damper}})
+ m_{\text{exciter}}\,\psi_1^2(l_{\text{exciter}})
+ m_{\text{absorber}}\,\psi_1^2(l_{\text{exciter}}).
$$

The effective stiffness of the reduced system is taken as

$$
k_a = k_{\text{beam}} = m_{m_1}\,\omega_{m_1}^2,
$$

where $\omega_{m_1}$ represents the first natural frequency of the beam without
the absorber, and $m_{m_1}$ is the corresponding first-mode modal mass defined
earlier. The effective damping coefficient $c_a$ is obtained from the free
vibration response of the beam–absorber system.


Damping Absorber.png

### Absorber–Coupled System Modelling and Calibration

Following the procedure adopted in Part~2 of the laboratory work, the system damping
coefficient is estimated from experimental measurements and used for model calibration.
Accordingly, the equivalent damping constant is taken as 16.77 N·s/m.

When the interaction between the primary structure and the vibration absorber is included,
the governing equation of motion for the beam can be expressed as

$$
m_a \ddot{z} + c_a \dot{z} + k_a z
- k_{\text{absorber}}(y_2 - y_1)
=
Q(\Omega)\,\psi_1(l_{\text{exciter}})\sin(\Omega t).
$$

Since the displacement of the beam at the exciter location is given by

$$
y_1 = z(t)\,\psi_1(l_{\text{exciter}}),
$$

the equation of motion becomes

$$
m_a \ddot{z} + c_a \dot{z}
+ \bigl(k_a + k_{\text{absorber}}\psi_1^2(l_{\text{exciter}})\bigr)z
- k_{\text{absorber}}\psi_1(l_{\text{exciter}})\,y_2
=
Q(\Omega)\,\psi_1(l_{\text{exciter}})\sin(\Omega t).
$$

---

### Absorber Equation of Motion

Assuming that the absorber damping is negligible, the dynamic equation governing the
absorber motion is written as

$$
m_2 \ddot{y}_2 + k_{\text{absorber}}(y_2 - y_1) = 0.
$$

Substituting the beam displacement expression yields

$$
m_2 \ddot{y}_2 + k_{\text{absorber}}y_2
- k_{\text{absorber}}\psi_1(l_{\text{exciter}})\,z = 0.
$$

---

### Coupled Two-Degree-of-Freedom System

The coupled beam–absorber system can therefore be represented in matrix form as

$$
\underbrace{
\begin{bmatrix}
m_a & 0 \\
0 & m_2
\end{bmatrix}}_{M_A}
\begin{bmatrix}
\ddot{z} \\
\ddot{y}_2
\end{bmatrix}
+
\underbrace{
\begin{bmatrix}
c_a & 0 \\
0 & 0
\end{bmatrix}}_{C_A}
\begin{bmatrix}
\dot{z} \\
\dot{y}_2
\end{bmatrix}
+
\underbrace{
\begin{bmatrix}
k_a + k_{\text{absorber}}\psi_1^2(l_{\text{exciter}}) &
- k_{\text{absorber}}\psi_1(l_{\text{exciter}}) \\
- k_{\text{absorber}}\psi_1(l_{\text{exciter}}) &
k_{\text{absorber}}
\end{bmatrix}}_{K_A}
\begin{bmatrix}
z \\
y_2
\end{bmatrix}
=
\begin{bmatrix}
F_1^{(m)} \\
0
\end{bmatrix}.
$$

The first two natural frequencies of the coupled system are subsequently evaluated using
the above mass and stiffness matrices.


### Approach 2: Rigid Beam Modelling

As discussed previously, when the system response is dominated by the first vibration mode,
the rigid beam approximation provides improved accuracy compared to the Euler–Bernoulli
beam model. Therefore, the experimental setup is re-analysed using rigid beam theory for
comparison.

Let $\theta(t)$ denote the angular displacement of the rigid beam (refer to Figure~2), and
let $I_A$ represent the total mass moment of inertia of the complete assembly about the fixed
end. This is given by

$$
I_A = I_{\text{beam}} + I_{\text{spring}} + I_{\text{exciter}} + I_{\text{damper}} + I_{\text{absorber}}.
$$

The individual inertia contributions are expressed as

$$
I_{\text{beam}} =
\frac{1}{12} m_{\text{beam}} l_{\text{beam}}^2
+ m_{\text{beam}}\left(\frac{l_{\text{beam}}}{2}\right)^2
= \frac{1}{3} m_{\text{beam}} l_{\text{beam}}^2,
$$

$$
I_{\text{spring}} =
\left(\frac{m_{\text{spring}}}{3} + m_{\text{fixing}}\right) l_{\text{spring}}^2,
$$

$$
I_{\text{exciter}} = m_{\text{exciter}} l_{\text{exciter}}^2,
\qquad
I_{\text{damper}} = m_{\text{damper}} l_{\text{damper}}^2,
\qquad
I_{\text{absorber}} = m_{\text{absorber}} l_{\text{exciter}}^2.
$$

---

### Equation of Motion of the Rigid Beam

Including damping, stiffness, and the absorber interaction, the rotational equation of motion
of the beam is written as

$$
I_A \ddot{\theta}
+ c\,l_{\text{damper}}^2 \dot{\theta}
+ k_{\theta}\theta
- k_{\text{absorber}}(y_2 - y_1)
=
l_{\text{exciter}} Q(\Omega)\sin(\Omega t).
$$

Since the beam displacement at the exciter location is

$$
y_1 = \theta\,l_{\text{exciter}},
$$

the governing equation becomes

$$
I_A \ddot{\theta}
+ c\,l_{\text{damper}}^2 \dot{\theta}
+ \bigl(k_{\theta} + k_{\text{absorber}} l_{\text{exciter}}^2\bigr)\theta
- k_{\text{absorber}} l_{\text{exciter}} y_2
=
l_{\text{exciter}} Q(\Omega)\sin(\Omega t).
$$

---

### Absorber Equation of Motion

Assuming negligible damping in the absorber, its translational motion is governed by

$$
m_2 \ddot{y}_2 + k_{\text{absorber}}(y_2 - y_1) = 0,
$$

which, upon substitution of $y_1$, yields

$$
m_2 \ddot{y}_2 + k_{\text{absorber}}(y_2 - \theta\,l_{\text{exciter}}) = 0.
$$

---

### Coupled Two-Degree-of-Freedom System

The rigid beam–absorber system can finally be written in matrix form as

$$
\underbrace{
\begin{bmatrix}
I_A & 0 \\
0 & m_2
\end{bmatrix}}_{M_A}
\begin{bmatrix}
\ddot{\theta} \\
\ddot{y}_2
\end{bmatrix}
+
\underbrace{
\begin{bmatrix}
c\,l_{\text{damper}}^2 & 0 \\
0 & 0
\end{bmatrix}}_{C_A}
\begin{bmatrix}
\dot{\theta} \\
\dot{y}_2
\end{bmatrix}
+
\underbrace{
\begin{bmatrix}
k_{\theta} + k_{\text{absorber}} l_{\text{exciter}}^2
& -k_{\text{absorber}} l_{\text{exciter}} \\
-k_{\text{absorber}} l_{\text{exciter}}
& k_{\text{absorber}
}
\end{bmatrix}}_{K_A}
\begin{bmatrix}
\theta \\
y_2
\end{bmatrix}
=
\begin{bmatrix}
l_{\text{exciter}} Q(\Omega)\sin(\Omega t) \\
0
\end{bmatrix}.
$$

The first two natural frequencies of the coupled system are then obtained by solving the
corresponding eigenvalue problem.
