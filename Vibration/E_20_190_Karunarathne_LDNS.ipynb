1. Continuous System Description

For small-amplitude transverse vibrations of a slender beam, the dynamics can be accurately described using the Euler--Bernoulli beam theory. The governing partial differential equation (PDE) of the continuous system is

\begin{equation}
\rho A \frac{\partial^2 y}{\partial t^2}
+ c \frac{\partial y}{\partial t}
+ EI \frac{\partial^4 y}{\partial x^4}
+ P \frac{\partial^2 y}{\partial x^2}
= f(t,x)
\end{equation}


where  
$y(x,t)$ is the transverse displacement,  
$\rho$ is the material density,  
$A$ is the cross-sectional area,  
$E$ is Young’s modulus,  
$I$ is the second moment of area,  
$c$ is the distributed viscous damping per unit length,  
$P$ is the axial compressive force, and  
$f(t,x)$ is the external transverse load per unit length.



2. Boundary conditions


The beam is assumed to have one end free and the base elastically restrained in rotation by a spring. The corresponding boundary conditions are

\begin{align}
y(0,t) &= 0, \qquad \frac{\partial^3 y}{\partial x^3}(l_{beam},t) = 0 \\
EI\frac{\partial^2 y}{\partial x^2}(0,t) - k_{\theta}\frac{\partial y}{\partial x}(0,t) &= 0,
\qquad \frac{\partial^2 y}{\partial x^2}(l_{beam},t) = 0
\end{align}

where the rotational stiffness is defined as

\begin{equation}
k_{\theta} = k_{spring} l_{spring}^2
\end{equation}




3. Operator Formulation

Let $\mathfrak{F}$ denote the infinite-dimensional vector space of four-times differentiable functions defined on the interval $[0,L]$. Define the linear operator $H:\mathfrak{F} \rightarrow \mathfrak{F}$ as

\begin{equation}
H = \frac{EI}{\rho A} \frac{\partial^4}{\partial x^4}
\end{equation}

Using this operator, the equations of motion can be written compactly as

\begin{equation}
\frac{\partial^2 y}{\partial t^2}
+ C \frac{\partial y}{\partial t}
+ H y
= q(t,x)
\end{equation}

where $c = C \rho A$ and $f(t,x) = \rho A q(t,x)$.

The operator $H$ maps a transverse deflection shape to the acceleration caused purely by beam bending, subject to the specified boundary conditions.

4. Free Vibration and Eigenvalue Problem

Assume that $H$ has a countably infinite set of distinct positive eigenvalues $\{\omega_k^2\}_{k=1}^{\infty}$ with corresponding eigenfunctions $\{\psi_k(x)\}_{k=1}^{\infty}$, representing the mode shapes of the beam.

The eigenvalue problem is therefore

\begin{equation}
H \psi_k(x) = \omega_k^2 \psi_k(x)
\end{equation}

Substituting the definition of $H$ yields

\begin{align}
\frac{EI}{\rho A} \frac{d^4 \psi_k(x)}{dx^4} &= \omega_k^2 \psi_k(x) \\
\frac{d^4 \psi_k(x)}{dx^4} &= \frac{\rho A \omega_k^2}{EI} \psi_k(x) \\
\frac{d^4 \psi_k(x)}{dx^4} &= \beta_k^4 \psi_k(x)
\end{align}

where

\begin{equation}
\beta_k^4 = \frac{\rho A \omega_k^2}{EI}
\end{equation}


5. General Solution of the Mode Shape Equation

The general solution of the fourth-order differential equation is

\begin{equation}
\psi_k(x) =
A_k \cosh(\beta_k x)
+ B_k \sinh(\beta_k x)
+ C_k \cos(\beta_k x)
+ D_k \sin(\beta_k x)
\end{equation}

where $A_k$, $B_k$, $C_k$, and $D_k$ are constants determined from the boundary conditions.

6. Application of Boundary Conditions

Base-End Conditions

From $\psi_k(0) = 0$,

\begin{equation}
A_k + C_k = 0
\end{equation}

The first and second derivatives at $x=0$ are

\begin{align}
\psi_k'(0) &= \beta_k (B_k + D_k) \\
\psi_k''(0) &= \beta_k^2 (A_k - C_k)
\end{align}

Using $C_k = -A_k$, the rotational boundary condition becomes

\begin{align}
EI \beta_k^2 (A_k - C_k) - k_{\theta} \beta_k (B_k + D_k) &= 0 \\
2 EI \beta_k^2 A_k - k_{\theta} \beta_k (B_k + D_k) &= 0
\end{align}

\subsubsection*{Free-End Conditions}

At $x = l_{beam}$, the boundary conditions yield

\begin{align}
\psi_k''(l_{beam}) = 0 \Rightarrow \quad
& A_k\left[\cosh(\beta_k l_{beam}) + \cos(\beta_k l_{beam})\right]
+ B_k \sinh(\beta_k l_{beam})
- D_k \sin(\beta_k l_{beam}) = 0 \\
\psi_k'''(l_{beam}) = 0 \Rightarrow \quad
& A_k\left[\sinh(\beta_k l_{beam}) - \sin(\beta_k l_{beam})\right]
+ B_k \cosh(\beta_k l_{beam})
- D_k \cos(\beta_k l_{beam}) = 0
\end{align}


7. Characteristic Equation

The above three equations form a homogeneous linear system

\begin{equation}
\mathbf{H}(\beta_k)
\begin{bmatrix}
A_k \\
B_k \\
D_k
\end{bmatrix}
= \mathbf{0}
\end{equation}

where $\mathbf{H}(\beta_k) \in \mathbb{M}_{3 \times 3}$.

For non-trivial solutions to exist, the determinant must vanish:

\begin{equation}
\det \mathbf{H}(\beta_k) = 0
\end{equation}

The roots $\beta_k$ are obtained numerically, and the corresponding constants $A_k$, $B_k$, and $D_k$ are found by solving the homogeneous system.

```python

import numpy as np
from numpy.linalg import det
from scipy.linalg import svd
from scipy.linalg import eigh
from scipy.optimize import brentq
from scipy.integrate import odeint
from scipy.integrate import quad
import matplotlib.pyplot as plt
from IPython.display import HTML
```

```python
# ========= Characteristic Matrix Determinant =========
def characteristic_determinant(beta, l_beam, EI, k_theta):

    bL = beta * l_beam

    H_mat = np.array([
        [2 * EI * beta**2,      -k_theta * beta,     -k_theta * beta],
        [np.cosh(bL) + np.cos(bL), np.sinh(bL),      -np.sin(bL)],
        [np.sinh(bL) - np.sin(bL), np.cosh(bL),      -np.cos(bL)]
    ])

    return det(H_mat)


def compute_beta_roots(l_beam, EI, k_theta, n_modes=2, beta_max=50):
    beta_roots = []
    beta_scan = np.linspace(1e-4, beta_max, 5000)

    det_vals = [
        characteristic_determinant(b, l_beam, EI, k_theta)
        for b in beta_scan
    ]

    for i in range(len(beta_scan) - 1):
        if det_vals[i] * det_vals[i + 1] < 0:
            beta_root = brentq(
                characteristic_determinant,
                beta_scan[i],
                beta_scan[i + 1],
                args=(l_beam, EI, k_theta)
            )
            beta_roots.append(beta_root)

            if len(beta_roots) >= n_modes:
                break

    return np.array(beta_roots)


# ============ Mode Shape Coefficient Solver ============
def compute_mode_shape_coeffs(beta, l_beam, EI, k_theta):

    bL = beta * l_beam

    H_mat = np.array([
        [2 * EI * beta**2,      -k_theta * beta,     -k_theta * beta],
        [np.cosh(bL) + np.cos(bL), np.sinh(bL),      -np.sin(bL)],
        [np.sinh(bL) - np.sin(bL), np.cosh(bL),      -np.cos(bL)]
    ])

    # Null-space via SVD (right singular vector with smallest singular value)
    _, _, Vh = svd(H_mat)
    null_vec = Vh[-1, :]     # proportional to [A, B, D]

    A, B, D = null_vec
    C = -A                  # from boundary condition

    return A, B, C, D
```

8. Reduced-Order (2-DOF) Approximation

The first two eigenfunctions $\psi_1(x)$ and $\psi_2(x)$ corresponding to the lowest eigenvalues $\omega_1^2$ and $\omega_2^2$ represent the first two dominant vibration modes. Retaining only these two modes provides a reduced-order 2-DOF model that captures the dominant dynamics of the beam.


Then, the eigenvalues of $H$ are given by,

$$
\{\omega_k^2\}_{k=1}^\infty = \left\{\frac{EI}{\rho A}{\beta_k}^4\right\}_{k=1}^\infty
$$

and the eigenfunctions/mode shapes of $H$ are given by,

$$
\{\psi_k\}_{k=1}^\infty = \left\{A_k \cosh(\beta_k x) + B_k \sinh(\beta_k x) + C_k \cos(\beta_k x) + D_k \sin(\beta_k x)\right\}_{k=1}^\infty
$$

$A_k$, $B_k$, $C_k$ and $D_k$ to be found.

###Forced Vibration Response Using Modal Expansion

1. External Forcing Model

Assuming there are no base excitations, the external transverse load acting on the beam is modelled as a harmonic point force applied at $x = l_{\text{exciter}}$:

\begin{equation}
q(x,t) = Q \sin(\Omega t)\,\delta(x - l_{\text{exciter}})
\end{equation}

2. Modal Decomposition of the Governing Equation

The governing equation of motion of the beam is

\begin{equation}
\ddot{y} + C\dot{y} + Hy = q(x,t)
\end{equation}

Using modal expansion, the system decouples into a set of forced, damped single-degree-of-freedom oscillators for each mode $k$:

\begin{equation}
m_k \ddot{z}_k
+ 2\zeta_k \omega_k m_k \dot{z}_k
+ m_k \omega_k^2 z_k
=
\langle\langle \psi_k(x), q(x,t) \rangle\rangle
\end{equation}

where

\begin{equation}
\zeta_k = \frac{C}{2\omega_k},
\qquad
m_k = \int_0^{l_{\text{beam}}} \rho A \, \psi_k^2(x)\,dx
\end{equation}

The modal mass $m_k$ normalises the arbitrary scaling introduced by the choice of mode shape constants.

3. Steady-State Modal Response

Solving the decoupled modal equation for harmonic excitation yields the steady-state response

\begin{equation}
z_k(t) = \chi_k(\Omega)\cos(\Omega t + \phi_k(\Omega))
\end{equation}

where $\chi_k(\Omega)$ is the steady-state amplitude and $\phi_k(\Omega)$ is the phase lag of the $k^{\text{th}}$ modal coordinate.



4. Evaluation of the Modal Forcing Term

The inner product defining the modal forcing is

\begin{align*}
\langle\langle \psi_k(x), q(x,t) \rangle\rangle
&= \int_0^{l_{\text{beam}}} \psi_k(x) q(x,t)\,dx \\
&= \int_0^{l_{\text{beam}}} \psi_k(x) Q \sin(\Omega t)\,\delta(x - l_{\text{exciter}})\,dx \\
&= Q \sin(\Omega t)\psi_k(l_{\text{exciter}})
\end{align*}


5. Steady-State Amplitude and Phase

The steady-state amplitude of the $k^{\text{th}}$ mode is given by

\begin{align*}
\chi_k(\Omega)
&=
\left|
\frac{\langle\langle \psi_k(x), q(x)\rangle\rangle}
{m_k \sqrt{(\omega_k^2 - \Omega^2)^2 + 4\zeta_k^2\Omega^2\omega_k^2}}
\right| \\
&=
\left|
\frac{Q \psi_k(l_{\text{exciter}})}
{m_k \sqrt{(\omega_k^2 - \Omega^2)^2 + 4\zeta_k^2\Omega^2\omega_k^2}}
\right|
\end{align*}

The phase lag is

\begin{equation}
\phi_k(\Omega)
=
\arctan\!\left(
\frac{2\zeta_k\Omega\omega_k}{\omega_k^2 - \Omega^2}
\right)
\end{equation}

6. Infinite-Dimensional Solution

The full displacement field of the beam is therefore expressed as

\begin{equation}
y(t,x)
=
\sum_{k=1}^{\infty} z_k(t)\psi_k(x)
=
\sum_{k=1}^{\infty}
\chi_k(\Omega)\cos(\Omega t + \phi_k(\Omega))\psi_k(x)
\end{equation}


###Reduction to a 2-DOF Model with Localised Spring--Damper Attachment

7. Two-Mode Approximation

Retaining only the first two dominant modes, the displacement is approximated by

\begin{align}
y(t,x) \approx\;
& \chi_1(\Omega)\cos(\Omega t + \phi_1(\Omega))\psi_1(x) \nonumber\\
&+ \chi_2(\Omega)\cos(\Omega t + \phi_2(\Omega))\psi_2(x)
\end{align}

8. Baseline 2-DOF Beam Model

The reduced-order system in modal coordinates is

\begin{align*}
\underbrace{
\begin{bmatrix}
m_1 & 0 \\
0 & m_2
\end{bmatrix}}_{M_{\text{beam}}}
\begin{bmatrix}
\ddot{z}_1 \\ \ddot{z}_2
\end{bmatrix}
+
\underbrace{
\begin{bmatrix}
2\zeta_1\omega_1 m_1 & 0 \\
0 & 2\zeta_2\omega_2 m_2
\end{bmatrix}}_{C_{\text{beam}}}
\begin{bmatrix}
\dot{z}_1 \\ \dot{z}_2
\end{bmatrix}
+
\underbrace{
\begin{bmatrix}
\omega_1^2 m_1 & 0 \\
0 & \omega_2^2 m_2
\end{bmatrix}}_{K_{\text{beam}}}
\begin{bmatrix}
z_1 \\ z_2
\end{bmatrix}
=
\underbrace{
\begin{bmatrix}
Q\psi_1(l_{\text{exciter}})\sin(\Omega t) \\
Q\psi_2(l_{\text{exciter}})\sin(\Omega t)
\end{bmatrix}}_{F_{\text{exciter}}}
\end{align*}

###Inclusion of Localised Attachments

9. Effective Mass of Spring Attachment

According to Rayleigh’s theory, one-third of the spring mass contributes to the system inertia. Including fixing mass,

\begin{equation}
m_{\text{eff}} = \frac{1}{3}m_{\text{spring}} + m_{\text{fixed}}
\end{equation}

The transverse acceleration at the attachment point is

\begin{align}
\ddot{y}(l_{\text{spring}},t)
=
\ddot{z}_1(t)\psi_1(l_{\text{spring}})
+
\ddot{z}_2(t)\psi_2(l_{\text{spring}})
\end{align}

The resulting inertia force is

\begin{equation}
F_{\text{inertia}}(t)
=
m_{\text{eff}}\ddot{y}(l_{\text{spring}},t)
\end{equation}

Projecting onto modal coordinates yields

\begin{align}
\begin{bmatrix}
F^{(m)}_1 \\
F^{(m)}_2
\end{bmatrix}
=
M_{\text{spring}}
\begin{bmatrix}
\ddot{z}_1 \\
\ddot{z}_2
\end{bmatrix}
\end{align}

with

\begin{align}
M_{\text{spring}}
=
m_{\text{eff}}
\begin{bmatrix}
\psi_1^2(l_{\text{spring}}) &
\psi_1(l_{\text{spring}})\psi_2(l_{\text{spring}}) \\
\psi_1(l_{\text{spring}})\psi_2(l_{\text{spring}}) &
\psi_2^2(l_{\text{spring}})
\end{bmatrix}
\end{align}

Additional mass matrices for the damper and exciter are similarly defined as

\begin{align}
M_{\text{damper}} &=
m_{\text{damper}}
\begin{bmatrix}
\psi_1^2(l_{\text{damper}}) &
\psi_1(l_{\text{damper}})\psi_2(l_{\text{damper}}) \\
\psi_1(l_{\text{damper}})\psi_2(l_{\text{damper}}) &
\psi_2^2(l_{\text{damper}})
\end{bmatrix} \\[2mm]
M_{\text{exciter}} &=
m_{\text{exciter}}
\begin{bmatrix}
\psi_1^2(l_{\text{exciter}}) &
\psi_1(l_{\text{exciter}})\psi_2(l_{\text{exciter}}) \\
\psi_1(l_{\text{exciter}})\psi_2(l_{\text{exciter}}) &
\psi_2^2(l_{\text{exciter}})
\end{bmatrix}
\end{align}


10. Damper Attachment Contribution

The viscous damper force is

\begin{equation}
F_{\text{damper}}(t)
=
c_{\text{damper}}\dot{y}(l_{\text{damper}},t)
\end{equation}

Using the two-mode approximation,

\begin{equation}
\dot{y}(l_{\text{damper}},t)
=
\dot{z}_1(t)\psi_1(l_{\text{damper}})
+
\dot{z}_2(t)\psi_2(l_{\text{damper}})
\end{equation}

The modal damping forces are

\begin{align}
F^{(c)}_k(t)
=
c_{\text{damper}}
\left(
\dot{z}_1\psi_1(l_{\text{damper}})
+
\dot{z}_2\psi_2(l_{\text{damper}})
\right)\psi_k(l_{\text{damper}})
\end{align}

which yields

\begin{align}
\begin{bmatrix}
F^{(c)}_1 \\
F^{(c)}_2
\end{bmatrix}
=
C_{\text{damper}}
\begin{bmatrix}
\dot{z}_1 \\
\dot{z}_2
\end{bmatrix}
\end{align}

with

\begin{align}
C_{\text{damper}}
=
c_{\text{damper}}
\begin{bmatrix}
\psi_1^2(l_{\text{damper}}) &
\psi_1(l_{\text{damper}})\psi_2(l_{\text{damper}}) \\
\psi_1(l_{\text{damper}})\psi_2(l_{\text{damper}}) &
\psi_2^2(l_{\text{damper}})
\end{bmatrix}
\end{align}

11. Final Reduced-Order System

Including all inertial and damping contributions, the final 2-DOF system is

\begin{equation}
(\mathbf{M}_{\text{beam}} + \mathbf{M}_{\text{spring}} + \mathbf{M}_{\text{damper}} + \mathbf{M}_{\text{exciter}})
\mathbf{\ddot{z}}
+
(\mathbf{C}_{\text{beam}} + \mathbf{C}_{\text{damper}})
\mathbf{\dot{z}}
+
\mathbf{K}_{\text{beam}}
\mathbf{z}
=
\mathbf{F}_{\text{exciter}}
\end{equation}

where $\mathbf{M}, \mathbf{C}, \mathbf{K} \in \mathbb{M}_{2\times2}$ and $\mathbf{z}, \mathbf{F} \in \mathbb{R}^2$.


```python
# ================= Beam Properties =================
b = 0.025          # Breadth (m)
h = 0.01           # Thickness (m)
A = b*h            # Cross-sectional area (m^2)
m_beam = 1.65      # Mass of the beam (m)
l_beam = 0.815     # Beam length (m)
E = 2e+11          # Modulus of Elasticity (Pa)
I = b * h**3 /12   # Area Moment of Inertia (m^4)
rho = m_beam/(A*l_beam) # Density (kg/m^3)

# ================= Spring Properties ==============
k_spring = 3800   # Spring stiffness (N/m)
l_spring = 0.75   # Location of spring along beam (m)
m_spring = 0.388  # Spring mass (kg)
m_fixed = 0.009   # Fixed mass attached to spring (kg)

# Rotational spring at base
k_theta = k_spring * l_spring * l_spring

# ================= Damper Properties ==============
l_damper = 0.2    # Location of damper along beam (m)
m_damper = 0.53   # Mass associated with damper (kg)

# ================= Exciter Properties =============
m_exciter = 4.2   # Exciter mass (kg)
l_exciter = 0.4   # Exciter location along beam (m)
```

```python
# ================= Find first two beta values =================
betas = compute_beta_roots(l_beam, E * I, k_theta, n_modes=2)
beta1, beta2 = betas

print(f"beta1 = {beta1:.4f}, beta2 = {beta2:.4f}")


# ================= Mode shape function =================
def psi(x, beta, l_beam, EI, k_theta):
    A, B, C, D = compute_mode_shape_coeffs(beta, l_beam, EI, k_theta)
    return (
        A * np.cosh(beta * x)
        + B * np.sinh(beta * x)
        + C * np.cos(beta * x)
        + D * np.sin(beta * x)
    )


# ================= Evaluate mode shapes at specific locations =================
psi1_spring = psi(l_spring, beta1, l_beam, E * I, k_theta)
psi2_spring = psi(l_spring, beta2, l_beam, E * I, k_theta)

psi1_damper = psi(l_damper, beta1, l_beam, E * I, k_theta)
psi2_damper = psi(l_damper, beta2, l_beam, E * I, k_theta)

psi1_exciter = psi(l_exciter, beta1, l_beam, E * I, k_theta)
psi2_exciter = psi(l_exciter, beta2, l_beam, E * I, k_theta)

print(f"psi1(l_spring)  = {psi1_spring:.4f}, psi2(l_spring)  = {psi2_spring:.4f}")
print(f"psi1(l_damper)  = {psi1_damper:.4f}, psi2(l_damper)  = {psi2_damper:.4f}")
print(f"psi1(l_exciter) = {psi1_exciter:.4f}, psi2(l_exciter) = {psi2_exciter:.4f}")


# ================= Modal mass calculation =================
def modal_mass(beta):
    """Modal mass = ∫ rho * A * psi^2 dx"""
    integrand = lambda x: rho * A * psi(x, beta, l_beam, E * I, k_theta)**2
    m, _ = quad(integrand, 0, l_beam)
    return m


m1 = modal_mass(beta1)
m2 = modal_mass(beta2)


# ================= Natural frequencies (beam only) =================
omega1 = beta1**2 * np.sqrt(E * I / (rho * A))
omega2 = beta2**2 * np.sqrt(E * I / (rho * A))

print(f"Modal mass m1 = {m1:.4f} kg, m2 = {m2:.4f} kg")
print(f"Natural frequencies (beam only):")
print(f"omega1 = {omega1:.4f} rad/s, omega2 = {omega2:.4f} rad/s")
```

```python
# ========= BEAM MATRICES =========
M_beam = np.array([
    [m1, 0.0],
    [0.0, m2]
])

K_beam = np.array([
    [m1 * omega1**2, 0.0],
    [0.0, m2 * omega2**2]
])


# ========= SPRING EFFECTIVE MASS =========
m_eff_spring = (1.0 / 3.0) * m_spring + m_fixed

M_spring = m_eff_spring * np.array([
    [psi1_spring**2, psi1_spring * psi2_spring],
    [psi1_spring * psi2_spring, psi2_spring**2]
])


# ========= DAMPER MASS =========
M_damper = m_damper * np.array([
    [psi1_damper**2, psi1_damper * psi2_damper],
    [psi1_damper * psi2_damper, psi2_damper**2]
])


# ========= EXCITER MASS =========
M_exciter = m_exciter * np.array([
    [psi1_exciter**2, psi1_exciter * psi2_exciter],
    [psi1_exciter * psi2_exciter, psi2_exciter**2]
])


# ========= TOTAL MASS MATRIX =========
M_total = M_beam + M_spring + M_damper + M_exciter


# ========= MATRIX PRINTER =========
def print_matrix(name, mat):
    print(f"{name}:")
    for row in mat:
        print("  [" + ", ".join(f"{val:10.3f}" for val in row) + "]")
    print()


# ========= OUTPUT =========
print_matrix("M_beam", M_beam)
print_matrix("M_spring", M_spring)
print_matrix("M_damper", M_damper)
print_matrix("M_exciter", M_exciter)
print_matrix("M_total", M_total)
print_matrix("K_beam", K_beam)
```

## Activity 2: Estimation of Damping Ratios for the Two Dominant Modes

We begin by recalling the governing equations of motion for the reduced-order system with the attached components, given by

$$
M \ddot{z} + C \dot{z} + K z = F_{\text{exciter}},
$$

where the system matrices are defined as

$$
M = M_{\text{beam}} + M_{\text{spring}} + M_{\text{damper}} + M_{\text{exciter}}, \quad
C = C_{\text{beam}} + C_{\text{damper}}, \quad
K = K_{\text{beam}}.
$$

To examine the inherent dynamic characteristics of the system, the free vibration response is considered by neglecting the external excitation. The corresponding homogeneous equation of motion is therefore

$$
M \ddot{z} + C \dot{z} + K z = 0.
$$

Assuming a harmonic solution of the form

$$
z(t) = \phi e^{i \omega t},
$$

and initially neglecting damping effects, the undamped response is used to determine the natural frequencies and mode shapes of the system. Substituting the assumed solution into the homogeneous equation yields the generalized eigenvalue problem

$$
(K - \omega^2 M)\phi = 0,
$$

with the associated characteristic equation

$$
\det(K - \omega^2 M) = 0.
$$

Solving this eigenvalue problem results in two natural frequencies, denoted by \( \omega_1 \) and \( \omega_2 \), together with their corresponding eigenvectors \( \phi_1 \) and \( \phi_2 \), where \( \phi \in \mathbb{R}^2 \).

The modal matrix is then constructed as

$$
\Phi =
\begin{bmatrix}
\phi_1 & \phi_2
\end{bmatrix}.
$$

The numerical evaluation of these quantities is carried out using the assembled mass and stiffness matrices defined previously.

```python
from scipy.linalg import eigh
import numpy as np

# -------------------------------------------------
# Assemble total mass and stiffness matrices
# -------------------------------------------------
M_total = M_beam + M_spring + M_damper + M_exciter
K_total = K_beam

# -------------------------------------------------
# Solve generalized eigenvalue problem
#   K Φ = ω² M Φ
# -------------------------------------------------
eigen_vals, eigen_vecs = eigh(K_total, M_total)

# Sort eigenvalues and eigenvectors (ascending order)
idx = np.argsort(eigen_vals)
eigen_vals = eigen_vals[idx]
eigen_vecs = eigen_vecs[:, idx]

# Natural frequencies
natural_freqs = np.sqrt(eigen_vals)

# Modal matrix
Phi = eigen_vecs

# -------------------------------------------------
# Output results
# -------------------------------------------------
print("Natural frequencies of complete system:")

for i, w in enumerate(natural_freqs, start=1):
    print(f"Mode {i}: {w:.3f} rad/s  =  {w/(2*np.pi):.3f} Hz")

print("\nSystem modal matrix Φ:")
print(Phi)
```

## Modal Transformation and Damping Ratio Estimation

A new set of generalized coordinates \( q(t) \) is introduced through the modal transformation

$$
z(t) = \Phi q(t),
$$

where \( \Phi \) is the modal matrix composed of the eigenvectors of the undamped system.

Substituting this transformation into the forced vibration equation yields

$$
M \Phi \ddot{q} + C \Phi \dot{q} + K \Phi q = F_{\text{exciter}}.
$$

Premultiplying by \( \Phi^{T} \), the equations of motion in modal coordinates are obtained as

$$
\Phi^{T} M \Phi \ddot{q}
+ \Phi^{T} C \Phi \dot{q}
+ \Phi^{T} K \Phi q
=
\Phi^{T} F_{\text{exciter}}.
$$

The modal mass, damping, and stiffness matrices are defined as

$$
M_m = \Phi^{T} M \Phi, \quad
C_m = \Phi^{T} C \Phi, \quad
K_m = \Phi^{T} K \Phi,
$$

and the corresponding modal force vector is

$$
F_m = \Phi^{T} F_{\text{exciter}}.
$$

The governing equations in modal coordinates therefore reduce to

$$
M_m \ddot{q} + C_m \dot{q} + K_m q = F_m.
$$

For the undamped eigenvectors of the generalized eigenvalue problem \( (K, M) \), the modal mass and stiffness matrices \( M_m \) and \( K_m \) are diagonal.

---

### Free Vibration in Modal Coordinates

Considering free vibration of the system, the modal equation becomes

$$
M_m \ddot{q} + C_m \dot{q} + K_m q = 0.
$$

In the modal formulation, each mode behaves as an independent single-degree-of-freedom oscillator provided that the damping matrix does not introduce modal coupling, i.e., when the modal damping matrix \( C_m \) is also diagonal.

---

### Modal Damping Ratio

The damping ratio of the \( k^{\text{th}} \) mode is defined as

$$
\zeta_k = \frac{c_k}{2 m_k \omega_k}.
$$

Using modal quantities, the damping ratio of the complete system can be expressed as

$$
\zeta_k
=
\frac{\phi_k^{T} C \phi_k}
{2 \omega_k \phi_k^{T} M \phi_k}.
$$

---

### Construction of the Damping Matrix

The total damping matrix of the system is given by

$$
C = C_{\text{beam}} + C_{\text{damper}}.
$$

The beam damping matrix \( C_{\text{beam}} \) is known to be diagonal as a result of the modal reduction of the beam alone. The contribution due to the viscous damper attachment is given by

$$
C_{\text{damper}}
=
c_{\text{damper}}
\begin{bmatrix}
\psi_1^2(l_{\text{damper}}) &
\psi_1(l_{\text{damper}})\psi_2(l_{\text{damper}}) \\
\psi_1(l_{\text{damper}})\psi_2(l_{\text{damper}}) &
\psi_2^2(l_{\text{damper}})
\end{bmatrix}.
$$

The analysis therefore proceeds by considering the mode shapes of the beam alone to evaluate the modal damping contributions.

\(l_{\text{damper}} = 0.2~\mathrm{m}\), the first mode shape is much smaller
in magnitude compared to the second mode shape:

$$
|\psi_1(l_{\text{damper}})| \ll |\psi_2(l_{\text{damper}})|.
$$

Specifically, we have already computed:

$$
\psi_1(l_{\text{damper}}) = -0.2491, \quad
\psi_2(l_{\text{damper}}) = 0.7102.
$$

From this, the relevant terms in the damper modal matrix are:

$$
\psi_2^2(l_{\text{damper}}) = 0.5044, \quad
\psi_1(l_{\text{damper}})\psi_2(l_{\text{damper}}) = -0.1769.
$$

Since

$$
|\psi_1(l_{\text{damper}})| \ll |\psi_2(l_{\text{damper}})|,
$$

the off-diagonal terms of \(C_{\text{damper}}\) are small and can be neglected.
This implies that the modal coupling introduced by the damper is negligible.

Assume **Rayleigh proportional damping** for the system:

$$
\mathbf{C} = \alpha\,\mathbf{M} + \beta\,\mathbf{K}
$$

Then, for the \(k\)-th mode:

$$
\phi_k^T \mathbf{C} \phi_k = \alpha\, \phi_k^T \mathbf{M} \phi_k + \beta\, \phi_k^T \mathbf{K} \phi_k \tag{3}
$$

Recall the eigenvalue relationship:

\begin{align*}
(\mathbf{K} - \omega_k^2 \mathbf{M}) \phi_k &= 0 \\
\phi_k^T \mathbf{K} \phi_k &= \omega_k^2 \, \phi_k^T \mathbf{M} \phi_k
\end{align*}

Substitute this into equation \((3)\):

\begin{align*}
\phi_k^T \mathbf{C} \phi_k &= \alpha\, \phi_k^T \mathbf{M} \phi_k + \beta\, (\omega_k^2 \phi_k^T \mathbf{M} \phi_k) \\
\phi_k^T \mathbf{C} \phi_k &= (\alpha + \beta\, \omega_k^2) \, \phi_k^T \mathbf{M} \phi_k
\end{align*}

Substituting into the expression for modal damping ratio:

$$
\zeta_k = \frac{\phi_k^T \mathbf{C} \phi_k}{2 \omega_k \, \phi_k^T \mathbf{M} \phi_k}
$$

we get:

$$
\zeta_k = \frac{(\alpha + \beta \, \omega_k^2) \, \phi_k^T \mathbf{M} \phi_k}{2 \omega_k \, \phi_k^T \mathbf{M} \phi_k}
$$

which simplifies to:

$$
\zeta_k = \frac{1}{2} \left( \frac{\alpha}{\omega_k} + \beta \, \omega_k \right)
$$

Since $|\psi_1(l_{beam})| << |\psi_2(l_{beam})|$, the damper is closer to the second mode antinode and the first mode node. This means it predominantly affects the second mode, that is the high frequency mode.

Therefore, we can assume stiffness proportional damping for this system ($\zeta_2 > \zeta_1$). Take $\alpha = 0$.

$$
\zeta_k = \frac{\beta \omega_k}{2}
$$


Since the maximum operating frequency of the exciter in the experimental setup is 17 Hz and $\omega_2$ > 17 Hz, we can only experimentally estimate $\zeta_1$ at $\omega_1$. Using that relationship, $\zeta_2$ can be found.

$$
\zeta_1 = \frac{\beta \omega_1}{2} \:\:\:\:\:\:\:\ \text{and} \:\:\:\:\:\:\:\ \zeta_2 = \frac{\beta \omega_2}{2}
$$

$$
\zeta_2 = \zeta_1 \frac{\omega_2}{\omega_1}
$$

Undamped Natural Frequency.png

The beam is initially displaced at its free end, so the resulting free-decay motion is largely dominated by the first vibration mode, which has its maximum amplitude (antinode) at the free end. As a result, the damping ratio extracted from the observed decay primarily represents the first modal damping, denoted as $\zeta_1$.

Assuming proportional damping, the damping ratio for the second mode can be approximated using the ratio of natural frequencies:

$$
\zeta_2 = \zeta_1 \frac{\omega_2}{\omega_1}.
$$

Inserting the given natural frequencies:

$$
\zeta_2 = 0.008 \times \frac{204.048~\text{rad/s}}{32.472~\text{rad/s}},
$$

which gives

$$
\zeta_2 \approx 0.05.
$$

Hence, the estimated damping ratios for the first two significant vibration modes are obtained as shown.

**3.1 Unforced Response Due to an Initial Displacement at the Free End of the Beam**

Starting from the modal coordinates of the full system, the physical displacement vector can be expressed as

$$
z(t) = \Phi q(t),
$$

or equivalently,

$$
\begin{bmatrix}
z_1(t) \\
z_2(t)
\end{bmatrix}
=
\begin{bmatrix}
\phi_{11} & \phi_{12} \\
\phi_{21} & \phi_{22}
\end{bmatrix}
\begin{bmatrix}
q_1(t) \\
q_2(t)
\end{bmatrix}.
$$

Recalling the truncated two-degree-of-freedom approximation of the beam displacement,

$$
y(x,t) = z_1(t)\psi_1(x) + z_2(t)\psi_2(x),
$$

the displacement field can be rewritten to include the effects of the attachments:

$$
y(x,t)
=
(\phi_{11} q_1(t) + \phi_{12} q_2(t))\psi_1(x)
+
(\phi_{21} q_1(t) + \phi_{22} q_2(t))\psi_2(x).
$$

Rearranging terms gives

$$
y(x,t)
=
(\phi_{11}\psi_1(x) + \phi_{21}\psi_2(x))q_1(t)
+
(\phi_{12}\psi_1(x) + \phi_{22}\psi_2(x))q_2(t),
$$

where the effective mode shapes of the complete system are defined as

$$
\psi_1^{*}(x) = \phi_{11}\psi_1(x) + \phi_{21}\psi_2(x), \qquad
\psi_2^{*}(x) = \phi_{12}\psi_1(x) + \phi_{22}\psi_2(x).
$$

If the initial displacement is applied at the beam tip:

$$
y(l_{\text{beam}}, 0) = y_0,
$$

we obtain

$$
y_0 = \psi_1^{*}(l_{\text{beam}})q_1(0) + \psi_2^{*}(l_{\text{beam}})q_2(0).
$$

Since there are two unknown modal coordinates and the initial displacement is applied at the free end, the response is assumed to be dominated by the first mode. Therefore, the initial conditions reduce to

$$
q_2(0) = 0,
\qquad
q_1(0) = \frac{y_0}{\psi_1^{*}(l_{\text{beam}})}.
$$

The time histories of \(q_1(t)\) and \(q_2(t)\) are obtained by solving the unforced modal equation:

$$
M_m \ddot{q} + C_m \dot{q} + K_m q = 0,
$$

which can be written in expanded form for proportional damping as

$$
M_m \ddot{q} + \beta K_m \dot{q} + K_m q = 0,
$$

with initial conditions

$$
q(0) =
\begin{bmatrix}
q_1(0) \\
0
\end{bmatrix},
\qquad
\dot{q}(0) =
\begin{bmatrix}
0 \\
0
\end{bmatrix}.
$$

Here, the damping coefficient \(\beta\) is given by

$$
\beta = \frac{2\zeta_1}{\omega_1} = \frac{2\zeta_2}{\omega_2}.
$$

Finally, the damped free-vibration response of the beam is reconstructed as

$$
y(x,t) = \psi_1^{*}(x)q_1(t) + \psi_2^{*}(x)q_2(t).
$$

```python
# ==================== Initial Displacement ====================
y_o = 0.003  # Prescribed tip displacement (meters)

# Modal matrix elements
phi11 = Phi[0, 0]
phi12 = Phi[0, 1]
phi21 = Phi[1, 0]
phi22 = Phi[1, 1]

# ================= Mode Shapes for Complete System =================
def psi_star_1(x):
    return phi11 * psi(x, beta1, l_beam, E*I, k_theta) + \
           phi21 * psi(x, beta2, l_beam, E*I, k_theta)

def psi_star_2(x):
    return phi12 * psi(x, beta1, l_beam, E*I, k_theta) + \
           phi22 * psi(x, beta2, l_beam, E*I, k_theta)

# Evaluate first effective mode shape at beam tip
psi_star_1_l_beam = psi_star_1(l_beam)
print(f"psi_star_1(l_beam) = {psi_star_1_l_beam:.4f}")

# Initial modal coordinate for first mode
q1_0 = y_o / psi_star_1_l_beam
print(f"Initial modal coordinate q1_0 = {q1_0:.4f}")

# ================== Diagonalized Matrices ======================
M_m = Phi.T @ M_total @ Phi      # Modal mass matrix
K_m = Phi.T @ K_total @ Phi      # Modal stiffness matrix

# ===================== Modal Damping ===========================
zeta1 = 0.008  # First mode damping ratio
zeta2 = 0.05   # Second mode damping ratio

# Rayleigh proportional damping coefficient (using first mode frequency)
beta_rayleigh = 2 * zeta1 / natural_freqs[0]
C_m = beta_rayleigh * K_m       # Modal damping matrix
```

```python
from scipy.integrate import odeint
import numpy as np
import matplotlib.pyplot as plt

# ================= ODE function for Free Vibration =================
def system_ode(q_vec, t, M_m_inv, C_m, K_m):
    q1, q2, q1_dot, q2_dot = q_vec
    q = np.array([q1, q2])
    q_dot = np.array([q1_dot, q2_dot])

    # Compute modal accelerations
    q_ddot = M_m_inv @ (-C_m @ q_dot - K_m @ q)

    return [q1_dot, q2_dot, q_ddot[0], q_ddot[1]]


# ================= Initial Conditions =================
M_m_inv = np.linalg.inv(M_m)

q1_0_val = q1_0  # Initial modal coordinate for first mode
q2_0_val = 0.0   # Second mode starts at zero

q_dot1_0 = 0.0
q_dot2_0 = 0.0

q_initial = [q1_0_val, q2_0_val, q_dot1_0, q_dot2_0]

# ================= Time Vector =================
t = np.arange(0, 10, 0.001)  # 10 seconds with 1 ms step

# ================= Solve ODE =================
solution = odeint(system_ode, q_initial, t, args=(M_m_inv, C_m, K_m))

# Modal responses
q1h_t = solution[:, 0]
q2h_t = solution[:, 1]

# ================= Transform to Physical Coordinates =================
y_l_beam_t = np.zeros_like(t)

psi1_l_beam = psi(l_beam, beta1, l_beam, E*I, k_theta)
psi2_l_beam = psi(l_beam, beta2, l_beam, E*I, k_theta)

for i in range(len(t)):
    q_vector = np.array([q1h_t[i], q2h_t[i]])
    z_vector = Phi @ q_vector
    # Free end displacement using first two mode shapes
    y_l_beam_t[i] = z_vector[0] * psi1_l_beam + z_vector[1] * psi2_l_beam

# ================= Plot Free End Response =================
plt.figure(figsize=(10, 3))
plt.plot(t, y_l_beam_t)
plt.xlabel('Time (s)')
plt.ylabel('Displacement (m)')
plt.title('Unforced Response at the Free End of the Beam')
plt.grid(True)
plt.tight_layout()
plt.show()
```

**3.2 Forced Response for Three Selected Motor Rotation Speeds**

We now analyze the forced vibration response of the reduced-order system.  
The equations of motion in modal coordinates are given by

$$
M_m \ddot{q} + C_m \dot{q} + K_m q = F_m,
$$

where the modal forcing vector is defined as

$$
F_m = \Phi^{T} F_{\text{exciter}},
$$

and the physical excitation force applied by the motor is expressed as

$$
F_{\text{exciter}} =
\begin{bmatrix}
Q(\Omega)\psi_1(l_{\text{exciter}})\sin(\Omega t) \\
Q(\Omega)\psi_2(l_{\text{exciter}})\sin(\Omega t)
\end{bmatrix}.
$$

The amplitude of the excitation due to the motor's rotating unbalance is

$$
Q(\Omega) = m_{\text{ecc}}\, r\, \Omega^{2},
$$

where $$m_{\mathrm{ecc}}$$ is the eccentric mass, and $$r$$ is the radial offset of the eccentric mass associated with the motor exciter.

The resulting system of non-homogeneous ordinary differential equations can be solved to obtain the particular solution \(q(t)\) corresponding to the harmonic forcing.  

Once the modal responses are obtained, the physical displacement of the beam can be reconstructed for plotting using

$$
y(x,t) = \psi_1^{*}(x)q_1(t) + \psi_2^{*}(x)q_2(t).
$$

```python
from scipy.integrate import odeint
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
import numpy as np
import matplotlib.pyplot as plt

# ================== Input Exciter Parameters ==================
Omega_1 = 32.472    # Near first resonance
Omega_2 = 100.0     # Intermediate
Omega_3 = 204.048   # Near second resonance

Omega_list = [Omega_1, Omega_2, Omega_3]

m_ecc =  0.0045     # Eccentric mass x 2 (kg)
r     =  0.035      # Offset (m)

# ================== Compute Excitation Force Amplitude ==========
def compute_Q(m_ecc, r, Omega):
    return 2 * m_ecc * r * Omega**2

# ================== ODE for Forced Vibration ==================
def system_ode_forced(q_vec, t, M_m_inv, C_m, K_m, Omega, Q, psi_star_l):
    q1, q2, q1_dot, q2_dot = q_vec
    q = np.array([q1, q2])
    q_dot = np.array([q1_dot, q2_dot])

    # Physical forcing vector
    f_physical = psi_star_l * Q * np.sin(Omega * t)

    # Forced equation of motion
    q_ddot = M_m_inv @ (f_physical - C_m @ q_dot - K_m @ q)

    return [q1_dot, q2_dot, q_ddot[0], q_ddot[1]]

# ================== Initial Conditions ==================
q_initial = [0.0, 0.0, 0.0, 0.0]
psi_star_l = np.array([psi_star_1(l_exciter), psi_star_2(l_exciter)])

# ================== Solve ODE for Each Excitation Frequency ==================
solutions = []

for Omega in Omega_list:
    Q = compute_Q(m_ecc, r, Omega)
    solution = odeint(
        system_ode_forced, q_initial, t,
        args=(M_m_inv, C_m, K_m, Omega, Q, psi_star_l)
    )
    solutions.append(solution)

# ================== Animate Forced Response ==================
def animate_forced_response(solution, Omega):
    q1_t = solution[:, 0]
    q2_t = solution[:, 1]

    y_x_t = np.zeros((len(t), len(x)))
    for i in range(len(t)):
        y_x_t[i, :] = psi_star_1_full(x) * q1_t[i] + psi_star_2_full(x) * q2_t[i]

    fig, ax = plt.subplots(figsize=(7, 3))
    ax.set_xlim(0, l_beam)
    max_disp = np.max(np.abs(y_x_t)) * 1.1
    ax.set_ylim(-max_disp, max_disp)
    ax.set_xlabel('Position along Beam (m)')
    ax.set_ylabel('Displacement (m)')
    ax.set_title(f'Forced Response, Ω = {Omega:.2f} rad/s')
    ax.grid(True)

    line, = ax.plot([], [], lw=2)

    def init():
        line.set_data([], [])
        return line,

    def animate(i):
        line.set_data(x, y_x_t[i, :])
        return line,

    skip = 10
    frames_idx = range(0, len(t), skip)

    animation = FuncAnimation(
        fig, animate, init_func=init, frames=frames_idx,
        interval=int((t[1] - t[0]) * skip * 1000),
        blit=True
    )

    plt.close(fig)
    return HTML(animation.to_html5_video())

# ================== Display Animations ==================
for idx, Omega in enumerate(Omega_list):
    display(animate_forced_response(solutions[idx], Omega))
```

**3.3 Theoretical Frequency Response of the System Capturing the First Two Dominant Modes**

The frequency response function (FRF) of the system is defined as the ratio of the steady-state displacement amplitude to the applied force amplitude:

$$
G(x,\Omega) = \frac{Y(x,\Omega)}{Q(\Omega)}.
$$

To account for the effects of the attachments, the modal amplitude and phase under harmonic excitation are expressed as

$$
\chi_k(\Omega)
=
\left|
\frac{Q(\Omega)\psi_k(l_{\text{exciter}})}
{M_{m,k}\sqrt{(\omega_k^2 - \Omega^2)^2 + 4\zeta_k^2\Omega^2\omega_k^2}}
\right|,
$$

and

$$
\phi_k(\Omega)
=
\arctan\!\left(
\frac{2\zeta_k\Omega\omega_k}{\omega_k^2 - \Omega^2}
\right),
$$

where \(\omega_k\) and \(\zeta_k\) are the natural frequencies and damping ratios of the complete system with attachments.

The physical displacement response can then be written in modal form as

$$
y(x,t) = \psi_1^{*}(x)q_1(t) + \psi_2^{*}(x)q_2(t),
$$

with the modal coordinates under harmonic excitation given by

$$
q_k(t) = \chi_k(\Omega)\cos\big(\Omega t + \phi_k(\Omega)\big).
$$

Substituting these into the displacement equation yields

$$
y(x,t)
=
\chi_1(\Omega)\cos(\Omega t + \phi_1(\Omega))\psi_1^{*}(x)
+
\chi_2(\Omega)\cos(\Omega t + \phi_2(\Omega))\psi_2^{*}(x).
$$

Using the identity
\(
\cos(\Omega t + \phi) = \Re\{e^{j(\Omega t + \phi)}\},
\)
the displacement can be expressed in complex form as

$$
y(x,t)
=
\Re\left\{
\left[
\chi_1(\Omega)\psi_1^{*}(x)e^{j\phi_1(\Omega)}
+
\chi_2(\Omega)\psi_2^{*}(x)e^{j\phi_2(\Omega)}
\right] e^{j\Omega t}
\right\}.
$$

Factoring out the time-dependent exponential identifies the frequency-domain displacement:

$$
Y(x,\Omega)
=
\chi_1(\Omega)\psi_1^{*}(x)e^{j\phi_1(\Omega)}
+
\chi_2(\Omega)\psi_2^{*}(x)e^{j\phi_2(\Omega)}.
$$

Explicitly incorporating magnitude and phase, the frequency-domain displacement can be written as

$$
Y(x,\Omega)
=
\frac{Q(\Omega)\psi_1(l_{\text{exciter}})\psi_1^{*}(x)}
{M_{m,1}\big[(\omega_1^2 - \Omega^2) + j2\zeta_1\Omega\omega_1\big]}
+
\frac{Q(\Omega)\psi_2(l_{\text{exciter}})\psi_2^{*}(x)}
{M_{m,2}\big[(\omega_2^2 - \Omega^2) + j2\zeta_2\Omega\omega_2\big]}.
$$

Finally, dividing by the force amplitude \(Q(\Omega)\), the frequency response function is obtained as

$$
G(x,\Omega)
=
\frac{\psi_1(l_{\text{exciter}})\psi_1^{*}(x)}
{M_{m,1}\big[(\omega_1^2 - \Omega^2) + j2\zeta_1\Omega\omega_1\big]}
+
\frac{\psi_2(l_{\text{exciter}})\psi_2^{*}(x)}
{M_{m,2}\big[(\omega_2^2 - \Omega^2) + j2\zeta_2\Omega\omega_2\big]}.
$$

The magnitude of \(G(x,\Omega)\) provides the displacement amplitude response, while the argument of \(G(x,\Omega)\) gives the corresponding phase response of the system.

```python
# ======================= Input Beam Positions for FRF =========================
x1 = l_beam
x2 = l_damper
x3 = 0.495       # Near node of second mode

x_vals_f = np.array([x1, x2, x3])

# Evaluate mode shapes at selected points
psi1_x = np.array([psi_star_1(xi) for xi in x_vals_f])
psi2_x = np.array([psi_star_2(xi) for xi in x_vals_f])

# Mode shapes at exciter location
psi1_le = psi_star_1(l_exciter)
psi2_le = psi_star_2(l_exciter)

# Natural frequencies and modal masses
omega_1, omega_2 = natural_freqs[0], natural_freqs[1]
M_m1, M_m2 = M_m[0, 0], M_m[1, 1]

# =========================== Evaluation of FRF ================================
def G(x_idx, Omega):
    denom1 = (omega_1**2 - Omega**2) + 1j * (2 * zeta1 * Omega * omega_1)
    denom2 = (omega_2**2 - Omega**2) + 1j * (2 * zeta2 * Omega * omega_2)

    term1 = (psi1_le * psi1_x[x_idx]) / (M_m1 * denom1)
    term2 = (psi2_le * psi2_x[x_idx]) / (M_m2 * denom2)

    return term1 + term2

# ============ Frequency Response at 3 Selected Positions on Beam ==============
Omega_vals = np.linspace(0.01, 300, 600)
G_all = np.zeros((3, len(Omega_vals)), dtype=complex)

for i in range(3):
    for j, Omega in enumerate(Omega_vals):
        G_all[i, j] = G(i, Omega)

# ======================== Plot FRF Magnitude and Phase ========================
fig, axs = plt.subplots(1, 2, figsize=(12, 5))

# Magnitude in dB
for i in range(3):
    axs[0].plot(Omega_vals, 20 * np.log10(np.abs(G_all[i])), label=f"$x_{i+1}$")
axs[0].set_xlabel(r"$\Omega$ [rad/s]")
axs[0].set_ylabel(r"$20\log_{10}|G(x,\Omega)|$ [dB]")
axs[0].set_title("FRF Magnitude (dB)")
axs[0].legend()
axs[0].grid(True)

# Phase (unwrap for clarity)
for i in range(3):
    axs[1].plot(Omega_vals, np.unwrap(np.angle(G_all[i])), label=f"$x_{i+1}$")
axs[1].set_xlabel(r"$\Omega$ [rad/s]")
axs[1].set_ylabel(r"Phase [rad]")
axs[1].set_title("FRF Phase")
axs[1].legend()
axs[1].grid(True)

plt.tight_layout()
plt.show()
```
